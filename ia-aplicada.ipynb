## 1.1 O Ciclo da Computação: Entrada, Processamento e Saída

**Ideias principais desta seção**

- **O Ciclo da Computação:** Todo processamento em computadores (seja um relógio, celular ou carro) segue um fluxo básico: os dados são inseridos (entrada), passam por um processamento e geram um resultado (saída).

Para compreender o universo da tecnologia e da inteligência artificial, é fundamental dominar os conceitos básicos de como as máquinas processam informações. Independentemente de estarmos falando de um relógio inteligente, um carro moderno, um celular ou um computador tradicional, todos eles compartilham a mesma essência: são computadores digitais projetados para a manipulação, análise e processamento de dados.

O processamento computacional segue um fluxo operacional básico e universal. Esse Ciclo da Computação é composto por três etapas principais: a entrada, o processamento e a saída.

- **Dados (Entrada):** O processo se inicia quando fornecemos os dados que desejamos que a máquina compute. No jargão da área de informática, frequentemente utiliza-se o verbo "entrar" com os dados (do inglês *enter*), caracterizando a fase de entrada de dados (*input*).
- **Algoritmo (Processamento):** Os dados inseridos são então submetidos a um algoritmo, que atuará como o motor de processamento desses dados.
- **Resultado (Saída):** Após a computação, a máquina devolve o produto final do processamento, que é chamado de resultado ou saída (*output*).

Visualmente, esse fluxo pode ser resumido de forma direta:


````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

etapas = ["DADOS\n(ENTRADA)", "ALGORITMO\n(INSTRUÇÕES)", "RESULTADO\n(SAÍDA)"]
cores = ["#118ab2", "#ffd166", "#06d6a0"]

fig, ax = plt.subplots(figsize=(10, 3))

largura, altura = 2.6, 1.4
espaco = 1.2
x = 0
fim_caixas = []

for etapa, cor in zip(etapas, cores):
    ax.add_patch(patches.FancyBboxPatch((x, 0), largura, altura,
                                         boxstyle="round,pad=0.05,rounding_size=0.15",
                                         linewidth=0, facecolor=cor))
    ax.text(x + largura / 2, altura / 2, etapa, ha="center", va="center",
            fontsize=13, fontweight="bold", color="#1a1a1a")
    fim_caixas.append(x + largura)
    x += largura + espaco

for fim in fim_caixas[:-1]:
    ax.annotate("", xy=(fim + espaco - 0.15, altura / 2), xytext=(fim + 0.15, altura / 2),
                arrowprops=dict(arrowstyle="-|>", color="#333333", lw=2.5))

ax.set_xlim(-0.3, x - espaco + 0.3)
ax.set_ylim(-0.3, altura + 0.5)
ax.axis("off")
ax.set_title("O Ciclo da Computação", fontsize=14, pad=15)
plt.tight_layout()
plt.show()
````

## 1.2 Algoritmos e Programas: da Receita de Bolo ao Código

**Ideias principais desta seção**

- **O que é um Algoritmo:** Trata-se de uma sequência de instruções ou comandos passo a passo para realizar uma tarefa específica, semelhante a uma receita de bolo.
- **Programa de Computador:** Quando essas instruções sequenciais são formalizadas usando regras gramaticais e uma sintaxe que o computador entende, temos um programa de computador.

Mas o que é um Algoritmo afinal? Embora muitas vezes seja confundido de forma abstrata com "filtros" ou "modelos matemáticos" que organizam perfis de usuários em redes sociais, a definição de algoritmo na base da ciência da computação é muito mais simples e direta. Um algoritmo é, essencialmente, uma sequência de etapas, comandos ou instruções passo a passo criadas para realizar uma determinada tarefa de forma geralmente sequencial (um passo após o outro). Essa noção de procedimento mecânico, que pode ser executado passo a passo sem exigir criatividade de quem executa, foi introduzida por Alan Turing ([Turing, 1936](https://doi.org/10.1112/plms/s2-42.1.230)).

Um exemplo clássico e didático para ilustrar um algoritmo é uma receita de bolo. Assim como uma receita orienta o cozinheiro sobre quais ingredientes usar (entrada) e qual a sequência exata de ações a tomar (processamento) para obter o bolo pronto (saída), o algoritmo orienta a máquina sobre como processar os dados recebidos.

A transição de um simples algoritmo, como uma receita ou uma ideia sequencial, para um programa de computador ocorre através da formalização da linguagem. Para que o computador compreenda as instruções a serem executadas, elas precisam ser escritas de acordo com uma sintaxe e regras gramaticais específicas de uma linguagem de programação.

Assim como as línguas naturais, como o português, inglês ou francês, possuem suas próprias regras gramaticais para que a comunicação as pessoas possam se comunicar naquele idioma, por exemplo sujeito, verbo e predicado, as linguagens de programação exigem regras objetivas e sem ambiguidade para que o computador consiga executar as instruções, o algoritmo. Quando um conjunto de instruções algorítmicas é escrito na sintaxe da linguagem de programação, seja em linguagens de programação de alto nível (ex. Python, Julia) ou baixo nível (Assembly), nós temos um programa de computador. Quando este programa é executado, abrindo um aplicativo por exemplo, o programa torna-se um **processo**, ou seja, um _programa em execução_ no celular, computador ou tablete.

## 1.3 A Comunicação com a Máquina: Sintaxe, *Bugs* e Níveis de Linguagem

**Ideias principais desta seção**

- **Sintaxe e Ambiguidade:** Assim como as línguas naturais, as linguagens de programação possuem gramática e sintaxe. A máquina não tolera ambiguidades: qualquer erro de comunicação com o computador pode gerar defeitos no programa, conhecidos como *bugs* (termo popularizado por Grace Hopper, que encontrou o inseto que causou um curto-circuito em um computador antigo).
- **Linguagem de Baixo Nível (A Línguagem da Máquina):** A máquina, em seu nível físico e de processador, só entende a linguagem binária (zeros e uns). Isso ocorre porque é tecnologicamente mais seguro mapear variações de tensão elétrica apenas como verdadeiro (1) ou falso (0) e vice-versa.
- **Linguagem de Alto Nível (Abstração):** Programar em zeros e uns é inviável para construirmos aplicativos complexos. Por isso, a computação criou camadas de abstração com linguagens de "alto nível" (como Python, SQL), que possuem uma sintaxe muito mais próxima da nossa linguagem natural, dessa maneira, facilitando a programação.

Comunicar-se é, em essência, conectar-se, interagir e transmitir informações para um receptor. No entanto, para que essa conexão seja bem sucedida, é necessário utilizar uma linguagem adequada ao seu público-alvo. Quando o nosso "público" é um computador, as regras dessa interação exigem protocolos de comunicação padronizados e rigorosos.

**Sintaxe, Ambiguidade e os "Bugs"**

Assim como as línguas naturais (como o português ou o inglês) exigem o conhecimento da gramática e de regras sintáticas para que a mensagem faça sentido (como a estrutura de sujeito, verbo e predicado), as linguagens de computador também operam sob regras estritas. Inclusive, as contribuições teóricas da linguística estrutural, como as de Noam Chomsky, foram fundamentais para modelar as linguagens de programação ([Chomsky, 1956](https://doi.org/10.1109/TIT.1956.1056813)).

A grande diferença na comunicação com a máquina é a sua intolerância à ambiguidade. Enquanto os humanos conseguem quebrar protocolos linguísticos na informalidade e ainda assim se entenderem, ser ambíguo com um processador resulta em prováveis falhas na execução das instruções. Essas falhas de comunicação e defeitos nos programas são conhecidos como *bugs* (insetos, em inglês). Apesar do [termo _bug_ ter sido criado por Thomas Edison](https://spectrum.ieee.org/did-you-know-edison-coined-the-term-bug), esse jargão popularizou-se graças a Grace Hopper, uma das mulheres pioneiras da programação e criadora do primeiro [compilador](https://pt.wikipedia.org/wiki/Compilador). Em 1947, a sua equipe encontrou uma mariposa presa na fiação de um computador Mark II quando buscavam o motivo de o computador não estar funcionando. Naquela época, os computadores eram enormes, podendo ocupar uma sala inteira. Elas descobriram que uma mariposa havia pousado nos fios, fechando um curto-circuito no computador e morreu no local. A partir de então, o termo passou a ser utilizado para se referir a defeitos nos programas. 

Vale registrar que a palavra *bug* já era usada por engenheiros para designar defeitos em equipamentos desde o século XIX. O episódio da mariposa, colado no relatório de uso do computador Harvard Mark II em 1947, foi o que popularizou o termo na computação ([Kidwell, 1998](https://doi.org/10.1109/85.728224)).

**A Língua da Máquina (Linguagem de Baixo Nível)**

Se quisermos descer ao nível físico e estrutural do processador, a verdadeira língua da máquina,  é o sistema binário: uma sequência muito grande de zeros e uns. Essa linguagem, categorizada como de "baixo nível", existe por uma necessidade de engenharia.

Os computadores processam dados à base de energia elétrica e precisam lidar com variações constantes de tensão (volts). Se os engenheiros tentassem mapear pequenas variações de voltagem para representar dezenas de caracteres, como letras do alfabeto (A a Z) e números (0 a 9), o projeto físico do computador ficaria complexo demais, muito caro e sujeito a erros constantes devido à instabilidade da energia elétrica. Para garantir precisão, o projeto foi simplificado: determina-se uma faixa de voltagem para representar o valor "1" (verdadeiro) e outra para representar o valor "0" (falso). Todo dado computado na máquina baseia-se apenas nesses dois valores. A base teórica para representar operações lógicas com circuitos elétricos de dois estados foi estabelecida por Claude Shannon ([Shannon, 1938](https://doi.org/10.1109/T-AIEE.1938.5057767)).

**A Abstração e as Linguagens de Alto Nível**

Naturalmente, escrever programas inteiros ditando ordens puramente em sequências de zeros e uns é um processo trabalhoso e, diante da complexidade dos softwares modernos, altamente inviável. Por exemplo, são necessários muias mais linhas de programação, instruções escritas em linguagem de programação, para programar um jogo de futebol em 3D do que uma calculadora que apenas soma. Para contornar essa dificuldade e facilitar o trabalho humano, a ciência da computação passou a criar camadas de abstração.

Essas camadas sobrepostas culminaram nas linguagens de alto nível. O adjetivo "alto nível" significa que a linguagem está muito mais próxima da forma como nós humanos nos comunicamos (nossa linguagem natural) do que da complexidade elétrica do processador. Linguagens de programação famosas, como o Python, possuem uma sintaxe amigável, permitindo que o programador escreva instruções com operadores matemáticos e variáveis simples. 

Suponha que queiramos escrever um programa que realiza a operação $(4 + 3) * 2$. O exemplo a seguir mostra um programa em linguagem de máquina, que implementa esse programa, com suas equivalentes conversões de Assembly para  hexadecimal e binário na arquitetura de computador x86):

| Instrução x86 | Hexadecimal | Código Binário (Máquina) | O que significa na prática? |
| :--- | :--- | :--- | :--- |
| `mov al, 4` | `B0 04` | **`10110000 00000100`** | `10110000` = Opcode (mover para AL). `00000100` = O número 4. |
| `add al, 3` | `04 03` | **`00000100 00000011`** | `00000100` = Opcode (somar valor em AL). `00000011` = O número 3. |
| `mov bl, 2` | `B3 02` | **`10110011 00000010`** | `10110011` = Opcode (mover para BL). `00000010` = O número 2. |
| `mul bl` | `F6 E3` | **`11110110 11100011`** | `11110110` = Opcode (multiplicação). `11100011` = Byte ModR/M que avisa que o alvo é o BL. |


Para escrevermos esse mesmo programa em uma linguage maior nível da abstração, isto é, mais perto da linguagem humana, utilizando a linguagem C ficaria assim:

```C
#include <stdio.h>

int main() {
    int resultado = (4 + 3) * 2;
    printf("Resultado: %d\n", resultado);
    return 0;
}
```

Já na linguagem SQL, o mesmo programa seria escrito da seguinte forma:

```sql
SELECT (4 + 3) * 2 AS resultado;
```

Em Julia, o programa seria esse a seguir:

```julia
resultado = (4 + 3) * 2
```

Escrevendo em linguagens de programação de mais alto nível de abstração, nós conseguimos instruir o computador de maneira fácil e o compilador se encarrega de converter tudo isso para o "baixo nível" binário que a máquina finalmente irá ler e executar. O **compilador** é um programa cuja função é converter programas de computador escritos em linguagens de programação de alto nível em linguagem de máquina, isto é, zeros e uns. O primeiro compilador foi descrito pela própria Grace Hopper ([Hopper, 1952](https://doi.org/10.1145/609784.609818)).

## 1.4 Funções e Programação em Linguagem Natural

**Ideias principais desta seção**

- **Aproximação com a Linguagem Natural:** É possível programar o computador usando comandos de voz comuns, como pedir ao celular: "Me acorde às 15:06".
- **Linguagem Matemática $f(x) = y$:** Essa programação corriqueira, "Me acorde às 15:06", é como a execução de uma função matemática pelo computador. Por exemplo, a funcionalidade (f) é o alarme, os parâmetros de entrada (x) é a hora e a saída (y) é o áudio reproduzido no horário escolhido.
- **Semiótica na Computação:** O uso de símbolos, letras e incógnitas serve para modelar os problemas em uma linguagem que intermedeia o pensamento humano e a instrução computacional.

A evolução das camadas de abstração na ciência da computação fez com que a comunicação com o computador atingisse níveis impressionantes de facilidade. Hoje, é perfeitamente possível programar um computador utilizando a nossa própria língua materna por exemplo. Quando você pega o seu celular e dá um comando de voz rotineiro, como "Me acorde às 15:06", você está programando a máquina. Esse tipo de interação, popularizada por assistentes virtuais em smartphones desde a década de 2010, traduz a nossa intenção falada diretamente para as instruções e ações que o sistema computacional precisa executar ([Hoy, 2018](https://doi.org/10.1080/02763869.2018.1404391)).

**O Poder de Abstração da Linguagem Matemática $f(x) = y$**

É muito comum na informática modularizarmos nosso programa em módulos menores que fazem ações específicas ([Parnas, 1972](https://dl.acm.org/doi/10.1145/361598.361623)), por exemplo um algortimo que verifica periodicamente se chegou nova mensagem no nosso celular. Uma forma de modularizar esses algoritmos é encapsulá-los em funções matemáticas. Voltando ao nosso exemplo cotidiano, embora ele pareça uma simples conversa informal, ajustar um alarme por comando de voz é o mesmo que executar a função matemática **$f(x) = y$**. Para desmistificar a utilização da Notação Matemática, aplicando ao exemplo do alarme, temos:

- O **$f$** representa a **função** ou funcionalidade do sistema que foi acionada (neste caso, alarme).
- O **$x$** representa o **parâmetro** da função, ou seja, os **dados de entrada** que você forneceu para processamento (o horário estipulado de 15:06).
- O **$y$** é o **resultado**, a **saída** final do processamento algorítmico, ou seja, o som reproduzido pontualmente às 15:06.

É importante notar que o nível de complexidade das funções pode variar. Existem funcionalidades que não exigem parâmetros, como dar o comando "Gere um número aleatório", então o computador simplesmente executa a ação e entrega a saída sem precisar de um dado extra de entrada. Por outro lado, existem funções com múltiplos parâmetros. No exemplo do alarme, poderíamos ter a variável $x_1$ como a hora desejada (15:06) e uma variável $x_2$ como a música específica a ser tocada como alarme. Parâmetros extras também podem ser opcionais, por exemplo caso você não informe o $x_2$ (a música), o algoritmo simplesmente seleciona o som padrão para cumprir a tarefa.

**Semiótica na Computação**

A utilização de letras, símbolos e incógnitas (como $f, x, y$) entra no campo da semiótica aplicada à computação ([Nadin, 1988](https://doi.org/10.1515/semi.1988.69.3-4.269)). Apesar de fórmulas matemáticas muitas vezes causarem certa aversão ou intimidação, elas nada mais são do que o uso de símbolos estruturados para modelar problemas reais.

Essa representação atua como uma linguagem simbólica vital, servindo de ponte intermediária entre o pensamento abstrato humano e as estruturadas instruções computacionais. Ao abstrairmos e modelarmos um problema do nosso dia a dia em variáveis lógicas, conseguimos estruturar a comunicação com a máquina de forma clara e objetiva, evitando a ambiguidade nas instruções fornecidas para o computador.

## 1.5 Inteligência da Máquina vs. Inteligência Humana

**Ideias principais desta seção**

- **A "Burrice" da Máquina:** O Computador não possui uma inteligência nos mesmos moldes do ser humano, pois tudo o que ele faz é processar instruções em linguagem de baixo nível. Deve-se ter empatia pela máquina, entendendo que ela é apenas uma ferramenta operando os dados que fornecemos.
- **O Diferencial Humano:** Um dos diferenciais entre a inteligência humana da computacional é a nossa capacidade de criar abstrações complexas (como famílias, bairros, sociedades) para resolver problemas e facilitar a convivência entre grandes grupos de indivíduos.

Quando observamos a impressionante capacidade de processamento dos computadores contemporâneos, é comum atribuirmos a eles características humanas. No entanto, é fundamental compreender as distinções entre a inteligência inerente ao ser humano, que é estudada por áreas como biologia, sociologia, psicologia e antropologia, e a forma como a máquina opera. É exatamente por causa dessa diferença conceitual que utilizamos o termo "inteligência artificial".

**A "Burrice" e a Empatia pela Máquina**

Apesar de sua capacidade de realizar cálculos supercomplexos em frações de segundo, a máquina, em sua essência, não é "inteligente" nos mesmos moldes que o cérebro humano. Tudo o que o computador faz é receber as instruções que escrevemos ou falamos, convertê-las para a sua linguagem de baixo nível (zeros e uns) e processá-las estritamente conforme foi comandado.

Nesse sentido, a máquina pode ser considerada desprovida de inteligência. Ela atua puramente como uma máquina de manusear números, operando às cegas os dados que nós mesmos fornecemos. Por isso, é necessário desenvolvermos o que se pode chamar de empatia pela máquina: quando um computador comete um erro ou executa uma ação indesejada, não faz sentido xingar o aparelho. A falha, na imensa maioria das vezes, reside na limitação de quem o programou, na ambiguidade do usuário que forneceu os comandos muto subjetivos ou, principalmente no caso da IA Generativa, na qualidade dos dados de treinamento.

Curiosamente, foi cientificamente provado que modelos de IA Generativa sob condições de estresse tendem a errar mais do que sob condições normais de estresse ([Goubin Shen et al., 2025](https://arxiv.org/abs/2409.17167)). Trata-se de um achado importantíssimo, mas que, sobretudo, não podemos entender como "a máquina tem sentimentos" e sim que ela tende a imitar o nosso comprtamento de errar mais sob condições de estresse. Por exemplo, colocando pressão em um prompt dizendo que o trabalho precisa ser feito porque a vida de pessoas depende daquilo, que o prazo já venceu, etc.

**O Diferencial Humano: A Criação de Abstrações**

Se a máquina é uma exímia executora sequencial de instruções, o que realmente define a inteligência humana e nos diferencia de outros animais? Uma das respostas está na nossa capacidade inata de criar abstrações.

Pesquisas científicas e sociológicas apontam que os seres humanos conseguem conviver e reconhecer organicamente os membros de uma comunidade de até aproximadamente 150 indivíduos ([Dunbar, 1992](https://doi.org/10.1016/0047-2484%2892%2990081-J)). Quando um grupo ultrapassa esse limite numérico, o reconhecimento mútuo se perde. Para resolver esse problema estrutural e conseguir organizar o convívio em sociedades gigantescas, a mente humana precisou inventar abstrações.

Conceitos como família, lares, bairros, ruas e a própria ideia de sociedade não existem no mundo físico como elementos físicos e tocáveis. Eles são abstrações complexas criadas pela nossa inteligência para facilitar a convivência e a organização de grandes grupos. 

Por outro lado, pode-se pedir para a IA criar abstrações também. Faça você mesmo um teste com alguma ferramenta como o Claude ([claude.ai](https://claude.ai)) pedindo para criar novas abstrações para nos ajudar na organização do ser humano em convívio em sociedade. Mas o que a Claude fará, independentemente de executar essa tarefa bem ou não, é reproduzir uma sequência de textos baseando-se na probabilidade de qual é a mais provável sequência de palavras que se encaixa com o comando que você deu. Inclusive, há críticos dos grandes modelos de linguagem (LLMs) que [referem-se às LLMs como papagaios](https://pt.wikipedia.org/wiki/Papagaio_estoc%C3%A1stico) que apenas geram textos sem ter a compreensão de fato da tarefa que está fazendo ([Bender et al., 2021](https://doi.org/10.1145/3442188.3445922)).

**A Inteligência como Forma de Resolver Problemas**

Acredito ser de suma importância lembrar-nos de que a inteligência artificial é apenas uma forma de resolver problemas como analisar dados e interpretar situações ([Newell & Simon, 1976](https://doi.org/10.1145/360018.360022)). Embora a inteligência do ser humano seja única em sua natureza biológica e social, não se descarta a possibilidade de que o avanço tecnológico permita que as máquinas, processando volumes insanos de dados, tornem-se capazes de criar abstrações computacionais ainda mais complexas do que aquelas que nós criamos ou que consigam resovler problemas complexos que nós nunca conseguiríamos. 

Ainda assim, a essência dessa "inteligência" continuará sendo artificial, operando como um reflexo algorítmico do conhecimento que nós a ensinamos a processar. A não ser que nos mesclamos com as máquinas e sejamos um ser só, [misto de humano com máquina](https://cochabambahotel.noblogs.org/files/2017/03/Manifesto_Ciborgue.pdf) ([Haraway, 1985](https://cochabambahotel.noblogs.org/files/2017/03/Manifesto_Ciborgue.pdf)), mas aí a discussão já extrapola o escopo da nossa discussão, pelo menos por enquanto.

## 1.6 O que é Inteligência Artificial Generativa

**Ideias principais desta seção**

- **O que é IA Generativa:** É um conjunto de algoritmos altamente complexos que utilizou bilhões de dados em sua fase de treinamento. Seu objetivo é simular a interação em linguagem natural e resolver problemas gerando novos dados.
- **Especialização e Limpeza de Dados:** A eficiência de uma IA depende muito dos dados que recebeu na sua fase de treinamento. Por exemplo, para ter uma IA voltada ao Direito, é preciso alimentá-la majoritariamente com dados jurídicos.

Para compreender como as modernas ferramentas de texto e imagem funcionam, precisamos desmistificar o que elas realmente são. A Inteligência Artificial Generativa não é mágica, mas sim uma coleção de algoritmos altamente complexos. O grande diferencial dessa tecnologia é que, durante a sua "fase de treinamento", esses algoritmos foram alimentados e processaram bilhões de dados. O objetivo central desse processamento massivo é capacitar a máquina a simular interações fluidas em linguagem natural com os seres humanos e utilizar essa própria linguagem para resolver problemas, gerando novos dados e respostas como resposta (saída). Na literatura científica, IA generativa é definida como um sistema computacional capaz de gerar conteúdo novo e aparentemente significativo, como texto, imagens ou áudio, a partir dos padrões aprendidos nos dados de treinamento ([Feuerriegel et al., 2024](https://doi.org/10.1007/s12599-023-00834-7)).

**Especialização vs. Inteligências Genéricas**

A eficiência e a utilidade de uma Inteligência Artificial estão diretamente relacionadas à qualidade e ao foco dos dados que ela ingeriu no treinamento. Se o objetivo de um desenvolvedor é criar uma IA voltada para o campo do Direito, ele precisará treiná-la majoritariamente com uma vasta quantidade de textos e processos jurídicos, diminuindo a carga de dados de outras áreas, como a psicologia. Dessa forma, a IA se torna altamente hábil em gerar conhecimentos e dados úteis para advogados e juízes. Um exemplo concreto é o LEGAL-BERT, um modelo treinado com centenas de milhares de documentos jurídicos que supera os modelos genéricos em tarefas do Direito ([Chalkidis et al., 2020](https://doi.org/10.18653/v1/2020.findings-emnlp.261)).

Em contrapartida, as IAs que utilizamos no dia a dia tendem a ser genéricas e interdisciplinares. Isso significa que foram treinadas com dados provenientes de uma ampla fonte de dados disponíveis na Internet para que saibam um pouco de tudo. O grande risco dessa abordagem abrangente é justamente a dificuldade de controle, pois ao absorver tudo de todos os lugares, a IA fica muito mais vulnerável a aprender e reproduzir vieses indesejados.

Esta seção explicou o que é a IA generativa e por que os dados de treinamento importam tanto. No Capítulo 2 vamos abrir o capô do modelo: como ele é treinado, quantos parâmetros tem e como prevê texto. No Capítulo 3 voltamos aos dados para entender a origem dos vieses (seção 3.1) e o que se pode fazer para mitigá-los (seção 3.2).

## 2.1 Do Aprendizado de Máquina ao Deep Learning

**Ideias principais desta seção**

- **Inteligência Artificial (IA):** campo de pesquisa iniciado por volta da década de 1950, com o objetivo de construir sistemas computacionais capazes de resolver problemas da forma como nós, seres humanos, resolvemos.
- **Aprendizado de Máquina:** uma das abordagens dentro da IA (ao lado de outras, como a programação lógica da linguagem Prolog) em que o sistema aprende a resolver problemas a partir de dados, em vez de seguir regras escritas manualmente por um programador.
- **Redes Neurais Artificiais:** uma forma de aprendizado de máquina inspirada em como o nosso cérebro funciona. Pegamos a ideia do neurônio biológico e a mapeamos em uma função matemática.
- **Deep Learning (Aprendizado Profundo):** o resultado de empilhar muitas camadas dessas redes neurais artificiais, uma sobre a outra.
- **Determinístico vs. probabilístico:** uma função comum devolve sempre a mesma saída para a mesma entrada; um modelo estatístico, não. O aprendizado de máquina, e portanto a IA generativa, é estatística: probabilístico por construção.

Antes de existir a IA generativa que usamos hoje, já existiam décadas de pesquisa em inteligência artificial buscando construir sistemas computacionais inteligentes. A ideia central, desde os anos 1950, sempre foi tentar resolver problemas da forma como nós, seres humanos, resolvemos: não necessariamente "raciocinar" como nós, mas produzir resultados equivalentes ([Turing, 1950](https://doi.org/10.1093/mind/LIX.236.433); [McCarthy et al., 2006](https://doi.org/10.1609/aimag.v27i4.1904)). Existem várias abordagens diferentes para tentar chegar lá. Uma delas é a programação lógica, exemplificada pela linguagem Prolog, que resolve problemas aplicando regras formais explícitas ([Colmerauer & Roussel, 1996](https://doi.org/10.1145/234286.1057820)). Outra, e é essa que nos interessa aqui, é o **aprendizado de máquina**, termo cunhado por Arthur Samuel ao descrever um programa que aprendia a jogar damas com a experiência ([Samuel, 1959](https://doi.org/10.1147/rd.33.0210)).

Dentro do aprendizado de máquina, existe uma família de técnicas chamada **redes neurais artificiais**. A inspiração é biológica: olhamos para como os neurônios do nosso cérebro se conectam e trocam sinais, e construímos funções matemáticas que tentam imitar esse comportamento. Cada "neurônio" artificial nada mais é do que uma função matemática: a mesma abstração $f(x) = y$ que vimos na seção 1.4, só que agora encadeada muitas vezes, em série e em paralelo ([McCulloch & Pitts, 1943](https://doi.org/10.1007/BF02478259); [Rosenblatt, 1958](https://doi.org/10.1037/h0042519)).

Com o passar do tempo, os pesquisadores começaram a empilhar cada vez mais desses neurônios, organizados em **camadas**. Quando uma rede neural tem muitas camadas, chamamos isso de **aprendizado de máquina profundo**, ou **deep learning** ([LeCun, Bengio & Hinton, 2015](https://doi.org/10.1038/nature14539)). Resumindo a hierarquia:

```
Inteligência Artificial
 └─ Aprendizado de Máquina
     └─ Redes Neurais Artificiais
         └─ Deep Learning (muitas camadas empilhadas)
```

Cada caixa está contida na anterior: todo deep learning é uma rede neural artificial, toda rede neural artificial é uma forma de aprendizado de máquina, e todo aprendizado de máquina é uma forma de inteligência artificial, mas o contrário não é verdade. A figura abaixo mostra essa hierarquia como conjuntos, um dentro do outro:

````python
# pip install matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Cada caixa está contida na anterior: todo deep learning é uma rede neural,
# toda rede neural é aprendizado de máquina, e todo aprendizado de máquina é IA.
niveis = [
    ("Inteligência Artificial", "#8ecae6"),
    ("Aprendizado de Máquina", "#219ebc"),
    ("Redes Neurais Artificiais", "#ffb703"),
    ("Deep Learning\n(muitas camadas empilhadas)", "#06d6a0"),
]

fig, ax = plt.subplots(figsize=(8, 6.5))

largura, altura = 10, 8
passo_x, passo_y = 0.9, 1.25   # cada caixa interna começa mais à direita e mais acima
for i, (nome, cor) in enumerate(niveis):
    x, y = i * passo_x, i * passo_y
    w, h = largura - 2 * i * passo_x, altura - i * passo_y - i * 0.35
    ax.add_patch(patches.FancyBboxPatch((x, y), w, h,
                                         boxstyle="round,pad=0.02,rounding_size=0.3",
                                         linewidth=1.5, edgecolor="white", facecolor=cor))
    if i < len(niveis) - 1:
        # o rótulo fica na faixa inferior, que a caixa interna não cobre
        ax.text(x + w / 2, y + passo_y / 2, nome, ha="center", va="center",
                fontsize=12, fontweight="bold", color="#1b1b1b")
    else:
        ax.text(x + w / 2, y + h / 2, nome, ha="center", va="center",
                fontsize=12, fontweight="bold", color="#1b1b1b")

ax.set_xlim(-0.3, largura + 0.3)
ax.set_ylim(-0.3, altura + 0.3)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Da Inteligência Artificial ao Deep Learning: uma hierarquia de conjuntos", fontsize=12, pad=12)
plt.tight_layout()
plt.show()
````

**Função determinística vs. modelo probabilístico**

Antes de seguir, vale entender uma diferença que explica boa parte do comportamento da IA generativa. As funções que vimos até aqui são **determinísticas**: para a mesma entrada, devolvem sempre a mesma saída. Considere a função $f(x) = 0{,}5x + 2$, em que $x$ é a quantidade de horas que uma pessoa estudou e $f(x)$ é a nota que ela vai tirar. Se $x = 5$, então $f(5) = 0{,}5 \cdot 5 + 2 = 4{,}5$. Pergunte de novo, e de novo: a resposta será sempre 4,5. É uma lei, altamente previsível, de certa forma estática. Não há variação.

Já um **modelo probabilístico** (ou estatístico) não funciona assim. Chame-o de $g(x)$. Para $x = 4$ horas de estudo, ele pode devolver, por exemplo, 4,8 em uma execução, 4,1 na seguinte e 3,9 na terceira. A variação pode ser grande ou pequena, mas ela existe: a saída oscila em torno de uma tendência e não se repete de forma exata. Em troca dessa perda de previsibilidade, o modelo probabilístico costuma **mapear melhor** os problemas do mundo real, que quase nunca obedecem a leis rígidas.

A estatística é uma ciência relativamente nova, com pouco mais de cem anos na forma como a conhecemos ([Stigler, 1986](https://www.hup.harvard.edu/books/9780674403413)). Diga-se de passagem que houve muito preconceito contra ela: os dogmáticos dentro da ciência (e é estranho falar de dogma em ciência) a ridicularizavam, até que seus métodos provaram ser úteis para resolver problemas que não conseguíamos resolver, ou que seriam complexos demais, com os métodos clássicos. Essa tensão entre a cultura dos modelos determinísticos e a dos modelos estatísticos foi descrita por Leo Breiman ([Breiman, 2001](https://doi.org/10.1214/ss/1009213726)). As abordagens probabilísticas, frequentistas e bayesianas, foram se desenvolvendo até conseguir resolver problemas complexos, e é sobre elas que se assenta o aprendizado de máquina: a base do aprendizado de máquina é a estatística ([Hastie, Tibshirani & Friedman, 2009](https://doi.org/10.1007/978-0-387-84858-7)). O aprendizado profundo é estatística; a IA generativa, construída sobre ele, também.

Guarde essa ideia, porque ela voltará várias vezes neste livro: **a IA generativa é probabilística por construção**. É isso que a torna poderosa e, ao mesmo tempo, o que a torna imprevisível. A figura abaixo compara as duas situações.

````python
# pip install matplotlib numpy
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)  # semente fixa: o "acaso" fica reproduzível
horas = np.arange(0, 11)        # 0, 1, 2, ..., 10 horas de estudo

# f(x): função determinística. Mesma entrada, mesma saída, sempre.
f = 0.5 * horas + 2

# g(x): modelo probabilístico. A saída oscila em torno da tendência.
# "Perguntamos" 3 vezes para cada quantidade de horas.
repeticoes = 3
g = np.array([0.5 * horas + 2 + rng.normal(0, 0.6, size=horas.size) for _ in range(repeticoes)])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.8), sharey=True)

# Esquerda: determinística
ax1.plot(horas, f, color="#219ebc", lw=2.5, label="f(x) = 0,5x + 2")
ax1.scatter([5], [4.5], color="#ef476f", zorder=3, s=90)
ax1.annotate("x = 5, três vezes:\n4,5   4,5   4,5", xy=(5, 4.5), xytext=(0.8, 6.6), fontsize=10,
             arrowprops=dict(arrowstyle="->", color="#555555"))
ax1.set_title("Função determinística")
ax1.set_xlabel("horas de estudo (x)")
ax1.set_ylabel("nota")
ax1.legend(loc="lower right")
ax1.grid(alpha=0.3)

# Direita: probabilística
ax2.plot(horas, f, color="#219ebc", lw=1.5, ls="--", label="tendência (0,5x + 2)")
for r in range(repeticoes):
    ax2.scatter(horas, g[r], color="darkorange", alpha=0.7, s=40,
                label="saídas de g(x)" if r == 0 else None)
ax2.scatter([4] * repeticoes, g[:, 4], color="#ef476f", zorder=3, s=90)
notas_4h = "   ".join(f"{v:.1f}".replace(".", ",") for v in g[:, 4])
ax2.annotate(f"x = 4, três vezes:\n{notas_4h}", xy=(4, g[:, 4].mean()), xytext=(0.3, 6.6), fontsize=10,
             arrowprops=dict(arrowstyle="->", color="#555555"))
ax2.set_title("Modelo probabilístico")
ax2.set_xlabel("horas de estudo (x)")
ax2.legend(loc="lower right")
ax2.grid(alpha=0.3)

fig.suptitle("Mesma entrada, mesma saída? Só na função determinística", fontsize=13)
plt.tight_layout()
plt.show()
````

Na seção 3.3 veremos que essa natureza probabilística está na raiz das alucinações; na seção 5.4, que ela impõe uma troca inevitável entre autonomia e controle quando a IA passa a executar tarefas.

## 2.2 Por que a IA Explodiu Agora: Algoritmos, GPUs e Bibliotecas

**Ideias principais desta seção**

- As redes neurais existem há décadas, mas só se tornaram competitivas quando três fatores se combinaram: **algoritmos melhores**, **GPUs acessíveis** e **bibliotecas** que facilitaram o processamento paralelo.
- Os marcos principais: nascimento da IA (1950), GPUs para ciência (2004) e a arquitetura CUDA (2008), Word2Vec (2013), redes generativas (2014), Transformers (2017) e o ChatGPT (2022).

Há cerca de 20 anos, as redes neurais artificiais eram consideradas o "patinho feio" da computação: uma abordagem que existia, mas que não conseguia resolver muitos problemas práticos de forma competitiva. Três fatores mudaram esse cenário e permitiram que o deep learning se popularizasse:

1. **Os algoritmos melhoraram.** Avanços teóricos tornaram o treinamento de redes com muitas camadas mais estável e mais eficaz ([Rumelhart, Hinton & Williams, 1986](https://doi.org/10.1038/323533a0); [Hinton, Osindero & Teh, 2006](https://doi.org/10.1162/neco.2006.18.7.1527)).
2. **O processamento em GPU ficou acessível.** GPU é a placa de vídeo do computador, o mesmo hardware feito originalmente para renderizar jogos. Há cerca de duas décadas, pesquisadores perceberam que podiam usar GPUs para processamento científico, porque saía mais barato do que comprar vários computadores tradicionais ([Buck et al., 2004](https://doi.org/10.1145/1015706.1015800); [Raina, Madhavan & Ng, 2009](https://doi.org/10.1145/1553374.1553486)). A NVIDIA, dona de boa parte desse mercado, inicialmente não permitia esse tipo de uso das suas placas. Ao perceber a oportunidade de negócio, a empresa mudou de postura: passou a permitir e a desenvolver software especificamente para viabilizar esse uso científico. A partir daí, a NVIDIA deixou de focar só em videogames e se tornou central no processamento de alto desempenho de forma geral: os computadores mais rápidos do mundo hoje usam GPUs, seja da própria NVIDIA ou de outras fabricantes.
3. **Bibliotecas de programação facilitaram o processamento paralelo.** Programar de forma distribuída dentro de uma GPU é, por natureza, uma tarefa difícil. O surgimento de bibliotecas que abstraem essa complexidade tornou essa técnica acessível a muito mais pesquisadores e desenvolvedores ([Nickolls et al., 2008](https://doi.org/10.1145/1365490.1365500)).

Com algoritmos melhores, hardware mais barato e ferramentas mais fáceis de usar, as redes neurais se mostraram muito boas para resolver problemas quando têm escala, isto é, muitos dados e muitas camadas. A tabela abaixo resume os principais marcos dessa trajetória:

## 2.3 Treinar vs. Usar um Modelo: Parâmetros, Custo e Água

**Ideias principais desta seção**

- **Treinar um modelo** significa dar dados para ele e deixar que o algoritmo encontre os números que fazem aquela função matemática funcionar bem: os **parâmetros** do modelo.
- **Parâmetros** são simplesmente os números de uma função. GPT e Gemini têm **centenas de bilhões ou mais** deles; as empresas não divulgam o número exato dos modelos comerciais, mas arquiteturas com mais de um trilhão de parâmetros já foram publicadas ([Fedus, Zoph & Shazeer, 2022](https://arxiv.org/abs/2101.03961)).
- **Treinar custa muito mais caro do que usar.** Treinar um modelo do porte do Claude, do GPT ou do Gemini pode custar centenas de milhões de dólares e demorar dias, entre energia elétrica, máquinas e água para resfriamento.
- **Modelos que rodam localmente no seu computador** costumam ser mais lentos e, principalmente, **mais limitados**: têm menos parâmetros e, portanto, menos capacidade de resolver problemas complexos.

Retome a abstração $f(x) = y$ da seção 1.4. Imagine que treinamos o modelo mais simples possível, uma função linear: $f(x) = ax + b$. Treinar esse modelo significa exatamente encontrar os valores de $a$ e $b$ que fazem a função descrever bem os dados que demos a ela. Esses dois números, $a$ e $b$, são os **parâmetros** do modelo. Esse modelo de exemplo tem apenas dois parâmetros.

**Um experimento em sala: horas de estudo e nota**

Vamos tornar isso concreto com um experimento feito em aula. Cada estudante é uma **observação**, uma medida do nosso conjunto de dados. Perguntamos a cada um: "Quantas horas você estudou para a prova? E que nota tirou?". Um estudou 5 horas e tirou 6; outro estudou 8 horas e tirou 9; e assim por diante. Colocando as horas no eixo horizontal e a nota no eixo vertical, cada estudante vira um ponto no gráfico. Esses pontos são os **dados de treinamento**.

O treinamento acontece em dois passos:

1. **Escolher o modelo.** Olhando para os pontos, decidimos que uma reta os representa bem. Poderia ser uma curva, e existem métodos para decidir qual modelo se ajusta melhor a um conjunto de dados ([Hastie, Tibshirani & Friedman, 2009](https://doi.org/10.1007/978-0-387-84858-7)), mas aqui, para fins didáticos, o olho basta. O modelo é $f(x) = ax + b$.
2. **Ajustar os parâmetros.** Um algoritmo de treinamento usa os dados coletados para encontrar os valores de $a$ e $b$ que deixam a reta o mais perto possível de todos os pontos ao mesmo tempo. Quando isso termina, dizemos que o modelo está **treinado**.

Suponha que o resultado foi $f(x) = 0{,}5x + 2$. O $2$ é onde a reta toca o eixo vertical; o $0{,}5$ diz que a reta não está a 45 graus, está "deitada" um pouco: cada hora a mais de estudo vale, em média, meio ponto. Com o modelo treinado, podemos fazer **previsões**: ninguém no nosso conjunto de dados estudou 9 horas, mas o modelo diz que quem estudar 9 horas deve tirar cerca de $0{,}5 \cdot 9 + 2 = 6{,}5$.

Repare em um detalhe que gera confusão em sala. O modelo diz que quem estuda 5 horas tira 4,5. "Mas, professor, o colega ali estudou 5 horas e tirou 6! Está errado." Não está: eu falei do modelo, não do ponto daquele estudante. O modelo **não é** o ponto de cada estudante, é a reta que resume a tendência de todos eles. Um modelo treinado representa o conjunto de dados de treinamento como um todo, não cada observação individualmente. A figura abaixo mostra os dados, a reta treinada e a previsão para 9 horas. Quem prefere a álgebra pode ficar com a equação; quem prefere ver, fica com o desenho.

````python
# pip install matplotlib numpy
import numpy as np
import matplotlib.pyplot as plt

# Dados de treinamento: cada par (horas, nota) é um estudante da turma.
horas = np.array([1, 2, 2, 3, 4, 4, 5, 5, 6, 7, 8, 8, 9.5, 10])
notas = np.array([2.5, 3.7, 2.3, 3.7, 3.1, 4.9, 6.0, 3.4, 3.6, 5.3, 9.0, 4.6, 5.6, 7.5])

# Passo 1: escolhemos o modelo, uma reta f(x) = a*x + b.
# Passo 2: o algoritmo de treinamento (aqui, mínimos quadrados) encontra a e b.
a, b = np.polyfit(horas, notas, deg=1)
print(f"Modelo treinado: f(x) = {a:.1f}x + {b:.1f}".replace(".", ","))

x = np.linspace(0, 10.5, 100)
fig, ax = plt.subplots(figsize=(8, 5.5))
ax.scatter(horas, notas, color="#1b1b1b", s=60, zorder=3, label="dados de treinamento (um ponto por estudante)")
ax.plot(x, a * x + b, color="#219ebc", lw=2.5,
        label=f"modelo treinado: f(x) = {a:.1f}x + {b:.1f}".replace(".", ","))

# Previsão para um caso que NÃO está nos dados: 9 horas de estudo.
x_novo = 9
y_novo = a * x_novo + b
ax.scatter([x_novo], [y_novo], color="#ef476f", s=180, marker="*", zorder=4,
           label=f"previsão para {x_novo} h de estudo: nota {y_novo:.1f}".replace(".", ","))
ax.vlines(x_novo, 0, y_novo, colors="#ef476f", ls=":", lw=1.5)
ax.hlines(y_novo, 0, x_novo, colors="#ef476f", ls=":", lw=1.5)

# O estudante que estudou 5 h e tirou 6 versus o que o MODELO diz para 5 h.
ax.annotate("um estudante: 5 h, nota 6", xy=(5, 6), xytext=(0.8, 8.4), fontsize=10,
            arrowprops=dict(arrowstyle="->", color="#555555"))
ax.annotate(f"o modelo: 5 h, nota {a * 5 + b:.1f}".replace(".", ","), xy=(5, a * 5 + b), xytext=(5.8, 2.9),
            fontsize=10, arrowprops=dict(arrowstyle="->", color="#555555"))

ax.set_xlim(0, 10.5)
ax.set_ylim(0, 10.5)
ax.set_xlabel("horas de estudo para a prova (x)")
ax.set_ylabel("nota na prova (y)")
ax.set_title("Treinar um modelo é encontrar a e b a partir dos dados")
ax.grid(alpha=0.3)
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
````

É a mesma coisa com um modelo de IA generativa: o texto usado no treinamento fica resumido nos seus parâmetros (a área também os chama de **pesos**), só que, em vez de dois números, são bilhões ou trilhões deles.

Os modelos de linguagem que usamos no dia a dia, como o ChatGPT, o Gemini ou o Claude, não têm dois parâmetros: eles têm **centenas de bilhões, possivelmente trilhões**. Cada um desses números foi ajustado durante o treinamento, a partir de bilhões de exemplos de texto. As empresas não divulgam o número exato de parâmetros dos modelos comerciais, mas arquiteturas com mais de um trilhão de parâmetros já foram descritas na literatura ([Fedus, Zoph & Shazeer, 2022](https://arxiv.org/abs/2101.03961)), e a pesquisa mostra que a capacidade do modelo cresce de forma previsível com o número de parâmetros, a quantidade de dados e o poder computacional usados no treinamento ([Kaplan et al., 2020](https://arxiv.org/abs/2001.08361)).

**O custo de treinar**

Treinar um modelo desse porte não é apertar um botão e esperar alguns minutos. É um processo que demora **dias**, exige **centenas de milhões de dólares** somando custo de energia elétrica e de máquinas, e consome **água** para resfriar os data centers onde o processamento acontece ([Strubell, Ganesh & McCallum, 2019](https://doi.org/10.18653/v1/P19-1355); [Patterson et al., 2022](https://doi.org/10.1109/MC.2022.3148714); [Li et al., 2025](https://doi.org/10.1145/3724499)). É por isso que passa tanto tempo entre o lançamento de uma versão de um modelo e a próxima: o treinamento é caro e lento. (Esse custo também está no centro do debate sobre modelos de negócio e uso de dados que veremos na seção 3.4.)

**Treinar vs. usar (inferência)**

Depois que um modelo é treinado, usá-lo (o que a área chama de **inferência**) também exige processamento intensivo e computadores potentes, mas em uma escala bem menor do que treinar:

| | Treinar | Usar (inferência) |
|---|---|---|
| **O que acontece** | O modelo ajusta seus parâmetros a partir dos dados | O modelo aplica os parâmetros já ajustados para gerar uma resposta |
| **Tempo** | Dias | Segundos |
| **Custo computacional** | Altíssimo | Bem menor, mas ainda relevante |
| **Frequência** | Acontece raramente (uma vez por versão do modelo) | Acontece a cada pergunta que você faz |

Vale uma ressalva: cada consulta individual custa pouco, mas, somadas as milhões de consultas feitas por dia, a energia gasta na inferência pode superar a do treinamento ao longo da vida útil do modelo ([de Vries, 2023](https://doi.org/10.1016/j.joule.2023.09.004)).

**Por que modelos locais são mais fracos**

Quando você baixa um modelo já treinado e o roda no seu próprio computador, ele tende a ser mais lento (porque o seu hardware é mais modesto do que um data center inteiro) e também **mais limitado do ponto de vista de capacidade**: geralmente são versões com muito menos parâmetros do que os modelos que rodam nos servidores das grandes empresas de tecnologia. Menos parâmetros significa, em geral, um modelo menos capaz de resolver problemas complexos ([Kaplan et al., 2020](https://arxiv.org/abs/2001.08361)).

## 2.4 O Papagaio Estatístico: Como o Modelo Prevê Texto

**Ideias principais desta seção**

- Um modelo de linguagem treinado é, essencialmente, um **preditor de texto**: dada uma sequência de palavras (o que você digitou), ele calcula qual é a próxima sequência de palavras com **maior probabilidade estatística** de fazer sentido ([Shannon, 1951](https://doi.org/10.1002/j.1538-7305.1951.tb01366.x); [Bengio et al., 2003](https://www.jmlr.org/papers/v3/bengio03a.html)).
- Antes de processar qualquer coisa, o computador converte o seu texto em uma **sequência de números**, porque, como vimos na seção 1.3, a máquina só entende zeros e uns.
- O modelo devolve outra sequência de números, que é convertida de volta em texto: a sua resposta.

Quando você manda uma pergunta para uma IA generativa, aquela sequência de caracteres que você digitou é convertida em uma sequência de números. Esse processo se chama **tokenização**. Cada pedaço de texto (uma palavra, ou parte dela) vira um número, chamado *token*. Os modelos atuais costumam quebrar palavras raras em pedaços menores, as subpalavras, para lidar com qualquer palavra, mesmo as que nunca viram ([Sennrich, Haddow & Birch, 2016](https://doi.org/10.18653/v1/P16-1162)). É assim que a máquina "lê" o que você escreveu, retomando o que vimos na seção 1.3: o computador só sabe fazer contas com números, porque no fim das contas tudo dentro dele é código binário.

Vamos ver como isso funciona na prática, convertendo um texto simples em tokens e números:


````python
# pip install huggingface_hub matplotlib
import re
from huggingface_hub import hf_hub_download
import matplotlib.pyplot as plt
import matplotlib.patches as patches

texto = "Eu vou para a classe, simbora?"

# mesmo vocabulário word2vec (NILC/USP) usado no exemplo mais à frente
vocab_path = hf_hub_download(repo_id="nilc-nlp/word2vec-skip-gram-50d", filename="vocab.txt")
with open(vocab_path, encoding="utf-8") as f:
    vocab = [w.strip() for w in f]
indice = {palavra: i for i, palavra in enumerate(vocab)}

# Simplificação didática: aqui cada palavra vira 1 token. Um tokenizador real de
# GPT quebra em subpalavras (ex.: "simbora" viraria "sim" + "b" + "ora").
tokens = re.findall(r"\w+|[^\w\s]", texto.lower())
ids = [indice[t] for t in tokens]

print(f"Texto original: {texto!r}")
print(f"Tokens (palavras): {tokens}")
print(f"Números (IDs no vocabulário word2vec): {ids}")

# --- Visualização: cada token vira uma caixinha colorida com o ID embaixo ---
cores = ["#ffd166", "#06d6a0", "#118ab2", "#ef476f", "#8338ec", "#fb5607", "#3a86ff", "#ff006e"]

fig, ax = plt.subplots(figsize=(11, 2.5))

x = 0
espaco = 0.15
for i, (token, id_) in enumerate(zip(tokens, ids)):
    largura = 0.35 * max(len(token), 1) + 0.3
    cor = cores[i % len(cores)]
    ax.add_patch(patches.FancyBboxPatch((x, 0), largura, 1,
                                         boxstyle="round,pad=0.02,rounding_size=0.08",
                                         linewidth=0, facecolor=cor))
    ax.text(x + largura / 2, 0.65, token, ha="center", va="center", fontsize=13, fontweight="bold")
    ax.text(x + largura / 2, 0.2, f"ID {id_}", ha="center", va="center", fontsize=9, color="#333333")
    x += largura + espaco

ax.set_xlim(-0.1, x)
ax.set_ylim(-0.3, 1.3)
ax.axis("off")
ax.set_title(f'Tokenização de "{texto}"', fontsize=13, pad=15)
plt.tight_layout()
plt.show()
````

Uma vez convertido em números, o modelo processa essa sequência e devolve **outra** sequência de números, que é traduzida de volta em texto. Esse texto de saída é escolhido porque, estatisticamente, é o que tem maior chance de fazer sentido em relação ao que você digitou.



**A analogia do teclado do celular**

Você já deve ter notado: quando digita "você vai" no teclado do celular, ele sugere palavras como *comer*, *ligar* ou *pagar*, e dificilmente sugere uma palavra aleatória sem nexo. Não é mágica: nos dados usados para treinar aquele teclado, essas palavras apareceram muito mais vezes logo depois de "você vai" do que qualquer outra alternativa. O modelo não "sabe" o que significa comer ou pagar. Ele apenas aprendeu que, estatisticamente, essas palavras costumam vir depois daquela sequência. É exatamente assim que funciona o teclado preditivo dos celulares, treinado com o histórico de digitação de milhões de usuários ([Hard et al., 2018](https://arxiv.org/abs/1811.03604)).

**O debate do papagaio**

Esse comportamento levou parte da comunidade de pesquisa a apelidar os grandes modelos de linguagem de **"papagaios estocásticos"**: sistemas que reproduzem sequências de palavras com base em probabilidade, sem ter, de fato, compreensão do que estão dizendo ([Bender & Koller, 2020](https://doi.org/10.18653/v1/2020.acl-main.463); [Bender et al., 2021](https://doi.org/10.1145/3442188.3445922)); voltamos a esse termo, já citado na seção 1.5. Há quem discorde e argumente que um modelo desse porte consegue produzir respostas que fazem sentido de verdade, mesmo sem ter "consciência" do conteúdo ([Mitchell & Krakauer, 2023](https://doi.org/10.1073/pnas.2215907120)). De um jeito ou de outro, um ponto é consensual: o modelo sabe muito de estatística. Ele não está entendendo o que você fala, mas sabe que, dada essa sequência de entrada, existe uma grande probabilidade de que uma determinada saída faça sentido.

**Palavras que "andam juntas" ficam parecidas**

Uma consequência direta desse mesmo princípio estatístico: palavras que aparecem com frequência perto umas das outras nos textos de treinamento acabam representadas por vetores parecidos, os chamados **embeddings**. Essa ideia, conhecida como hipótese distribucional, é anterior aos computadores modernos: Zellig Harris já propunha que o significado de uma palavra pode ser caracterizado pelas palavras que costumam acompanhá-la ([Harris, 1954](https://doi.org/10.1080/00437956.1954.11659520)). Isso vale até para palavras opostas em significado, desde que pertençam ao mesmo assunto. "Paz" e "guerra", por exemplo: são antônimos, mas aparecem o tempo todo nos mesmos tipos de texto: notícias, tratados, livros de história. Já uma palavra sem nenhuma relação com esse assunto, como "manjericão", quase nunca aparece perto de nenhuma das duas.

Vamos verificar isso com um exemplo real, usando **embeddings word2vec**, um dos primeiros modelos a aprender esse tipo de representação vetorial em larga escala, e que foi um marco importante na história do processamento de linguagem natural ([Mikolov et al., 2013](https://arxiv.org/abs/1301.3781)); voltamos a esse marco na seção 2.2. Os vetores usados abaixo foram treinados para o português pelo NILC/USP ([Hartmann et al., 2017](https://arxiv.org/abs/1708.06025)):


````python
%pip install -q huggingface_hub safetensors numpy scikit-learn matplotlib

from huggingface_hub import hf_hub_download
from safetensors.numpy import load_file
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import matplotlib.pyplot as plt

# Embeddings word2vec (skip-gram, 50 dim) pré-treinados pelo NILC/USP
# em 1,39 bilhão de tokens de português (fonte: huggingface.co/nilc-nlp)
vocab_path = hf_hub_download(repo_id="nilc-nlp/word2vec-skip-gram-50d", filename="vocab.txt")
emb_path = hf_hub_download(repo_id="nilc-nlp/word2vec-skip-gram-50d", filename="embeddings.safetensors")

with open(vocab_path, encoding="utf-8") as f:
    vocab = [w.strip() for w in f]
indice = {palavra: i for i, palavra in enumerate(vocab)}

matriz = load_file(emb_path)["embeddings"]  # (929606, 50)

palavras = ["manjericão", "paz", "guerra"]
vetores = np.stack([matriz[indice[p]] for p in palavras])  # (3, 50) - embeddings "de verdade"

print("Similaridade com embeddings reais (word2vec)\n")
similaridade = cosine_similarity(vetores)
print(f"'{palavras[0]}' x '{palavras[1]}': {similaridade[0][1]:.2f}")
print(f"'{palavras[0]}' x '{palavras[2]}': {similaridade[0][2]:.2f}")
print(f"'{palavras[1]}' x '{palavras[2]}': {similaridade[1][2]:.2f}")

# Normaliza cada vetor para comprimento 1 ANTES de reduzir a dimensão.
# Similaridade de cosseno só depende do ÂNGULO entre os vetores, não do
# tamanho deles. Normalizando, a seta no gráfico também vira só direção,
# e a distância da origem deixa de "confundir" a leitura do aluno.
vetores_unitarios = vetores / np.linalg.norm(vetores, axis=1, keepdims=True)

# TruncatedSVD (sem centralizar, diferente do PCA) preserva a origem (0, 0)
# como o vetor nulo de verdade, então o ângulo entre as setas no gráfico 2D
# continua se aproximando do ângulo real dos vetores em 50D.
svd = TruncatedSVD(n_components=2, random_state=42)
vetores_2d = svd.fit_transform(vetores_unitarios)

fig, ax = plt.subplots(figsize=(6, 6))

margem = 0.3
x_min, y_min = vetores_2d.min(axis=0) - margem
x_max, y_max = vetores_2d.max(axis=0) + margem
ax.set_xlim(min(x_min, 0), max(x_max, 0))
ax.set_ylim(min(y_min, 0), max(y_max, 0))

for palavra, (x, y) in zip(palavras, vetores_2d):
    ax.annotate("", xy=(x, y), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color="darkorange", lw=2))
    ax.text(x, y, f"  {palavra}", fontsize=12)

ax.axhline(0, color="gray", lw=0.5)
ax.axvline(0, color="gray", lw=0.5)
ax.set_aspect("equal")
ax.set_title("Embeddings word2vec (50D), normalizados e reduzidos para 2D")
ax.set_xlabel("dimensão 1")
ax.set_ylabel("dimensão 2")
plt.show()
````

Repare que "paz" e "guerra" ficaram com uma similaridade bem mais alta entre si do que qualquer uma delas com "manjericão": mesmo sendo opostas em significado, "paz" e "guerra" pertencem ao mesmo campo semântico (conflito) e aparecem juntas com frequência nos textos usados para treinar o modelo.

**Isso ajuda a explicar por que os modelos de linguagem "sabem" tanto sobre o mundo sem nunca terem "vivido" nada.** Eles não entendem o que é uma guerra nem o que é manjericão, mas, ao processar bilhões de palavras, aprendem quais palavras costumam aparecer perto de quais outras. Essa estatística de coocorrência acaba capturando, indiretamente, relações de significado, de assunto e até de oposição.

## 2.5 Modelo vs. Aplicativo: Onde Ficam os _Guardrails_

**Ideias principais desta seção**

- Um **modelo treinado**, sozinho, não é um produto pronto para uso: é a peça central sobre a qual se constroem **aplicativos**.
- O chat que usamos no dia a dia (ChatGPT, Claude, Gemini) é um desses aplicativos: um **assistente virtual construído em cima de um modelo treinado**.
- É na camada de **aplicativo**, e não no modelo em si, que ficam os filtros de palavrão, os limites de uso por plano de assinatura e os **guardrails** (proteções contra discurso de ódio e conteúdo não inclusivo).
- Os guardrails podem ser burlados por **injeção de prompt**: instruções disfarçadas que tentam se passar pelas regras do aplicativo. A defesa é um jogo de gato e rato.

Depois que um modelo é treinado, ele pode servir de base para **vários programas diferentes**, não só um chat. É possível construir, por exemplo, um assistente de atendimento, uma ferramenta de correção de texto ou um sistema de busca, todos usando o mesmo modelo por baixo. O chat, com o qual estamos tão acostumados, é apenas um desses aplicativos possíveis.

Visualizando essa arquitetura em camadas:

````python
# pip install matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as patches

etapas = ["Dados de\ntreinamento", "Treinamento", "Modelo\ntreinado", "Aplicativo\n(ex.: chat)", "Você"]
cores = ["#8ecae6", "#8ecae6", "#219ebc", "#ffb703", "#06d6a0"]

fig, ax = plt.subplots(figsize=(12, 4))

largura, altura = 1.8, 1.0
espaco = 0.9
y_base = 1.5

x = 0
centros = []
for i, (etapa, cor) in enumerate(zip(etapas, cores)):
    ax.add_patch(patches.FancyBboxPatch((x, y_base), largura, altura,
                                         boxstyle="round,pad=0.02,rounding_size=0.08",
                                         linewidth=0, facecolor=cor))
    ax.text(x + largura / 2, y_base + altura / 2, etapa, ha="center", va="center",
            fontsize=11, fontweight="bold", color="#1b1b1b")
    centros.append(x + largura / 2)
    if i < len(etapas) - 1:
        ax.annotate("", xy=(x + largura + espaco, y_base + altura / 2),
                     xytext=(x + largura, y_base + altura / 2),
                     arrowprops=dict(arrowstyle="-|>", color="#555555", lw=2))
    x += largura + espaco

# Caixa dos guardrails, abaixo do "Aplicativo": é aí, e não no modelo, que
# ficam os filtros de conteúdo e os limites de plano.
gx_centro = centros[3]
g_largura, g_altura = 3.6, 0.9
gx = gx_centro - g_largura / 2
gy = 0.0
ax.add_patch(patches.FancyBboxPatch((gx, gy), g_largura, g_altura,
                                     boxstyle="round,pad=0.02,rounding_size=0.08",
                                     linewidth=0, facecolor="#ef476f"))
ax.text(gx_centro, gy + g_altura / 2, "filtros de conteúdo, guardrails,\nlimites de plano",
        ha="center", va="center", fontsize=9.5, fontweight="bold", color="white")

ax.annotate("", xy=(gx_centro, y_base), xytext=(gx_centro, gy + g_altura),
             arrowprops=dict(arrowstyle="-|>", color="#ef476f", lw=2))

ax.set_xlim(-0.3, x - espaco + 0.3)
ax.set_ylim(-0.3, y_base + altura + 0.3)
ax.axis("off")
ax.set_title("Do treinamento até você: onde ficam os guardrails", fontsize=13, pad=12)
plt.tight_layout()
plt.show()
````

É dentro dessa camada de aplicativo que moram as regras de negócio e as proteções: se uma determinada palavra deve ser bloqueada, se um usuário do plano gratuito tem direito a menos mensagens por mês, se uma resposta potencialmente ofensiva deve ser reescrita ou recusada. Os **guard rails** (as proteções contra discurso de ódio e geração de conteúdo não inclusivo) são implementados justamente nesse nível, e não no modelo puro. Ferramentas como o NeMo Guardrails, da NVIDIA, existem exatamente para programar essas regras na camada de aplicativo ([Rebedea et al., 2023](https://doi.org/10.18653/v1/2023.emnlp-demo.40)).

Essa distinção tem uma consequência prática importante, que costuma passar despercebida: **o mesmo modelo pode se comportar de formas diferentes dependendo do aplicativo em que está rodando**, porque a camada que fica por cima dele muda. É também por isso que o comportamento de uma IA pode mudar de um dia para o outro: não porque o modelo foi retreinado, mas porque a empresa ajustou as regras do aplicativo. Pesquisadores documentaram mudanças grandes no comportamento do ChatGPT em poucos meses, tanto para melhor quanto para pior, sem que um novo modelo tivesse sido anunciado ([Chen, Zaharia & Zou, 2024](https://doi.org/10.1162/99608f92.5317da47)).

**Injeção de prompt: quando alguém tenta burlar os guardrails**

Se as regras do aplicativo são escritas em linguagem natural, o que impede um usuário de escrever, na própria conversa, uma instrução que finja ser uma dessas regras? É exatamente isso que se chama **injeção de prompt** (*prompt injection*): inserir, na troca de mensagens com o sistema, comandos que tentam se passar pelas instruções de quem programou ou personalizou a IA, para fazê-la executar algo que ela não foi configurada para fazer ([Perez & Ribeiro, 2022](https://arxiv.org/abs/2211.09527); [Liu, Deng et al., 2023](https://arxiv.org/abs/2306.05499)). A técnica pode ser usada de forma legítima, para testar o sistema, ou de forma maliciosa.

O cerne do problema é que esse tipo de tecnologia recebe instruções de **duas fontes**: de quem a configurou para o uso de terceiros (as regras do que pode ou não ser feito) e, naturalmente, dos próprios terceiros que a usam. Imagine uma ferramenta que faz slides. Quem a configurou passou instruções dizendo o que ela pode ou não fazer. Eu, ao pedir os meus slides, posso esconder dentro do meu comando instruções fingindo que sou a pessoa que preparou a ferramenta. Se ela não tiver as proteções adequadas, os guardrails, vai executar o comando.

A injeção pode vir até de um documento ou mensagem que a IA lê para responder, e não da pessoa que está perguntando ([Greshake et al., 2023](https://doi.org/10.1145/3605764.3623985)). Um caso clássico é o do assistente de IA do Slack: pesquisadores conseguiram, por injeção de prompt, fazer o assistente entregar dados de conversas e arquivos privados de outros usuários a pessoas que não tinham acesso a eles ([PromptArmor, 2024](https://promptarmor.substack.com/p/data-exfiltration-from-slack-ai-via)). Casos assim aconteceram várias vezes.

**Um jogo de gato e rato**

Por que os guardrails nem sempre funcionam? Porque a defesa é uma brincadeira de gato e rato. Faça de conta que uma equipe testou todos os modelos que existem, construiu um conjunto de testes (um *benchmark*) e chegou a proteções perfeitas contra injeção de prompt. Na semana seguinte, sai um modelo novo. A equipe roda o mesmo benchmark e descobre que o modelo novo consegue ser invadido. Vai ter que entender por quê e criar outra proteção. E o ciclo recomeça, porque o ritmo de lançamento dessas tecnologias é frenético. A literatura documenta exatamente isso: ataques que atravessam as proteções de vários modelos ao mesmo tempo ([Zou et al., 2023](https://arxiv.org/abs/2307.15043)) e falhas estruturais no treinamento de segurança que explicam por que novos ataques continuam surgindo ([Wei, Haghtalab & Steinhardt, 2023](https://arxiv.org/abs/2307.02483)).

Esse é o cotidiano da área de **segurança da informação**: sempre haverá uma técnica nova de proteção e, logo depois, uma técnica nova de ataque, possivelmente mais poderosa. Por isso a segurança é uma área tão pertinente na informática, e hoje, com a IA generativa, a preocupação é maior do que nunca. Voltaremos a esse padrão de gato e rato na seção 3.8, ao discutir a detecção de conteúdo gerado por IA.

## 3.1 A Origem dos Vieses: os Dados de Treinamento

**Ideias principais desta seção**

- **A Origem dos Vieses e Preconceitos:** Se uma Inteligência Artificial emite respostas racistas ou com discurso de ódio, a resposta mais provável é que os dados usados durante o seu treinamento continham este tipo de conteúdo. A IA Generativa gera conteúdo de texto, áudio ou vídeo influenciados pelas informações que "consumiu" durante a fase de treinamento.
- **Rotulação de Dados e Trabalho Humano:** por trás da limpeza e da rotulação dos dados de treinamento há um enorme esforço de pessoas, muitas vezes em condições precárias.

**A Origem dos Vieses e Preconceitos**

Como vimos anteriormente, a máquina executa instruções baseadas nos dados que recebe. Portanto, se uma IA generativa fornece uma resposta contendo discurso de ódio, racismo ou qualquer tipo de preconceito, isso não significa necessariamente que a máquina tenha desenvolvido malícia ou vontade própria. Comumente, isso ocorre porque os dados com os quais o algoritmo foi alimentado durante sua fase de treinamento já estavam enviesados e carregavam esses vieses. Dessa forma, a máquina atua como um espelho: ela simplesmente reflete e reproduz as características, falhas e padrões presentes nas informações que "consumiu". Isso foi demonstrado experimentalmente: representações de palavras aprendidas a partir de textos da internet reproduzem os mesmos preconceitos de gênero, raça e idade que testes psicológicos medem em seres humanos ([Caliskan, Bryson & Narayanan, 2017](https://doi.org/10.1126/science.aal4230); [Bolukbasi et al., 2016](https://arxiv.org/abs/1607.06520)).

**Limpeza, Rotulação de Dados e a Exploração Humana**

Garantir que os dados inseridos no algoritmo sejam de alta qualidade é um dos maiores desafios contemporâneos da ciência da computação. Para evitar que uma Inteligência Artificial comporte-se de forma indesejada, por exemplo retornando respostas equivocadas (alucinações; ver a seção 3.3 e ([Ji et al., 2023](https://doi.org/10.1145/3571730))), é necessário um rigoroso trabalho de preparação dos dados. Por exemplo, rotular os dados descrevendo que contém uma imagem ou dizendo se um texto contém discurso de ódio ou racista. 

Ironicamente, por trás da automação com IA para rotulação de dados, existe um gigantesco esforço de trabalho humano. Pessoas são utilizadas nos bastidores das grandes empresas de tecnologia para revisar, limpar e rotular manualmente esses dados de treinamento. Esse fator humano garante que a máquina seja influenciada apenas da forma esperada pelos seus criadores, mitigando respostas fora do objetivo. Infelizmente, essa tarefa muitas vezes é realizada de forma precária [por seres humanos](https://www.noemamag.com/the-exploited-labor-behind-artificial-intelligence/). Estudos sobre esse trabalho de "microtarefas", frequentemente terceirizado para países de baixa renda, mostram que ele é um componente estrutural, e não passageiro, da produção de IA ([Tubaro, Casilli & Coville, 2020](https://doi.org/10.1177/2053951720919776); [Miceli, Posada & Yang, 2022](https://doi.org/10.1145/3492853)), e que a história dos grandes conjuntos de dados, como o ImageNet, foi construída de modo a tornar esse trabalho invisível ([Denton et al., 2021](https://doi.org/10.1177/20539517211035955)).

Esta seção explicou a **origem** dos vieses. A próxima mostra como eles aparecem na prática e o que se pode fazer para reduzi-los.

## 3.2 Viés na Prática e Técnicas de Mitigação

**Ideias principais desta seção**

- O modelo devolve o padrão **mais frequente** nos dados de treinamento; por isso "um cientista" tende a virar um homem branco mais velho.
- Mitigar viés é possível **nos dados**, **no aplicativo** (reescrevendo o prompt) ou por **rodízio** de representações, mas enviesar sempre para o lado oposto também é enviesar.
- Viés não é só demográfico: também aparece em temas políticos e geopolíticos.

A seção 3.1 já explicou a **origem** dos vieses: eles vêm dos dados de treinamento. Esta seção trata de um passo adiante: **como esse viés aparece na prática** e **o que se pode fazer para reduzi-lo**.

**O exemplo do cientista**

Peça a uma IA generativa de imagens para "mostrar um cientista". É bem provável que o resultado seja um homem, branco e mais velho. O modelo não tem malícia: ele simplesmente reproduz o padrão mais frequente nos dados que recebeu como entrada. Se a maioria das imagens de "cientista" no conjunto de dados de treinamento mostrava esse perfil, essa é a resposta que carrega a maior probabilidade estatística, e é isso que o modelo devolve. Estudos com geradores de imagem populares confirmam e quantificam esse efeito: os modelos não apenas reproduzem, mas amplificam estereótipos de gênero e raça associados a profissões ([Bianchi et al., 2023](https://doi.org/10.1145/3593013.3594095); [Luccioni et al., 2023](https://arxiv.org/abs/2303.11408)).

**Técnicas de mitigação**

| Onde atua | Como funciona | Limitação |
|---|---|---|
| **Nos dados** | Selecionar dados reais mais diversos, ou gerar **dados sintéticos**: cientistas mulheres, negras, trans, jovens, sem jaleco branco, fora de laboratórios de química ou biologia | Exige trabalho deliberado de curadoria; não resolve o problema em modelos já treinados |
| **No aplicativo** | O aplicativo **reescreve o prompt** antes de enviá-lo ao modelo; por exemplo, "gere a imagem de uma cientista jovem, de origem indiana" ([Friedrich et al., 2023](https://arxiv.org/abs/2302.10893)) | Se aplicado sempre da mesma forma, cria um novo viés (ver abaixo) |
| **No rodízio** | Alternar entre diferentes representações a cada requisição, em vez de sempre reforçar a mesma imagem | Exige um mecanismo de aleatoriedade controlada; difícil de calibrar |

Vale destacar um ponto sutil sobre a segunda e a terceira técnica: **enviesar sempre para o lado oposto também é enviesar**. Se toda vez que alguém pedir "a imagem de um cientista" o sistema devolver sempre uma mulher jovem de determinada origem, isso não resolve o viés, apenas o desloca. O homem branco mais velho **é**, de fato, representativo dentro do universo real de cientistas; o problema não é ele aparecer, é ele ser a **única** representação possível. A solução mais equilibrada é o rodízio: variar as representações ao longo de várias requisições, sem excluir nenhuma delas.

**Viés não é só demográfico: também é político**

Um exemplo mais sutil de viés, levantado em aula, é a pergunta "o que é Taiwan?". Diferentes sistemas de IA já deram respostas incompatíveis entre si: alguns descreveram Taiwan como uma ilha independente e democrática; outros, como uma parte da China que se rebela contra o governo central. A divergência não vem de um "erro técnico": vem, de novo, dos dados de treinamento e de decisões editoriais de quem construiu cada modelo. Viés, portanto, não se limita a estereótipos de gênero, raça ou idade: também aparece em temas geopolíticos sensíveis. Pesquisas mostram que os vieses políticos estão presentes desde os dados de pré-treinamento e se propagam até as respostas finais ([Feng et al., 2023](https://doi.org/10.18653/v1/2023.acl-long.656)), e que o ChatGPT apresenta inclinação política sistemática, inclusive em perguntas sobre o Brasil ([Motoki, Pinho Neto & Rodrigues, 2024](https://doi.org/10.1007/s11127-023-01097-2)).

## 3.3 Alucinações: Quando o Modelo Inventa

**Ideias principais desta seção**

- "Alucinação" é o nome dado a uma resposta **errada** gerada pela IA. O modelo não mente: ele gera o que os dados de treinamento tornaram **mais provável**.
- Uma segunda causa de erro é a **memória de contexto**: em conversas longas, o modelo tende a esquecer, principalmente o que ficou no meio.
- Mitigação: **fornecer os dados** de onde a resposta deve sair, fazer **uma pergunta por vez**, **testar** o modelo na sua tarefa e consultar **benchmarks** feitos por especialistas.

**O que é uma alucinação**

Na seção 2.4 vimos que um modelo de linguagem é um preditor de texto: dada uma sequência, ele devolve a continuação mais provável. Chama-se **alucinação** o caso em que essa continuação é convincente e **falsa**: um artigo científico que não existe, uma lei com número inventado, uma estatística sem fonte, uma frase atribuída à pessoa errada ([Ji et al., 2023](https://doi.org/10.1145/3571730); [Huang et al., 2025](https://doi.org/10.1145/3703155); [Walters & Wilder, 2023](https://doi.org/10.1038/s41598-023-41032-5)). O termo é discutível, porque sugere que a máquina "vê coisas". Não é isso. Uma alucinação é simplesmente uma resposta gerada pela IA que está errada porque **houve maior probabilidade daquilo ser gerado**, devido aos dados de treinamento. Essa é uma das principais causas.

**O exemplo de Maria, João e José**

Vamos entender essa causa com um exemplo hipotético. Faça de conta que existiu uma pesquisadora chamada Maria, que tinha um assistente de pesquisa chamado João. Muitos textos usados no treinamento de um modelo eram trabalhos acadêmicos de Maria, e boa parte deles tinha João como coautor. Maria era casada com José, mas sobre a vida pessoal dela havia pouco texto na internet. A quantidade massiva de dados era sobre a sua atividade profissional. Logo, nos dados de treinamento, o nome "Maria" aparece junto de "João" muitas mais vezes do que junto de "José".

Depois de treinado, alguém pergunta ao modelo: "Maria era casada com João?". Há uma chance de ele dizer que sim. O modelo vai tentar se lembrar da relação entre os dois, e o que ele "lembra" é que trabalhavam juntos; talvez não lembre nada da vida pessoal de Maria. Um modelo mais esperto dirá que não sabe. Um modelo mais incipiente, como os de 2022 e 2023, poderia facilmente fazer a inferência errada de que Maria era casada com João, porque a probabilidade desses dois nomes estarem juntos era maior. O gráfico abaixo ilustra o desequilíbrio que está na origem do erro:

````python
# pip install matplotlib
import matplotlib.pyplot as plt

# Contagem FICTÍCIA de quantas vezes cada par de nomes aparece junto
# nos textos usados para treinar o modelo.
pares = ["Maria + João\n(coautor)", "Maria + José\n(marido)"]
ocorrencias = [340, 12]
cores = ["#219ebc", "#ffb703"]

fig, ax = plt.subplots(figsize=(7, 4.5))
barras = ax.bar(pares, ocorrencias, color=cores, width=0.55)
for barra, n in zip(barras, ocorrencias):
    ax.text(barra.get_x() + barra.get_width() / 2, n + 6, f"{n} vezes", ha="center", fontsize=11, fontweight="bold")

ax.set_ylabel("vezes que os nomes aparecem juntos\nnos dados de treinamento")
ax.set_ylim(0, 400)
ax.set_title("Pergunta: \"Com quem Maria era casada?\"\nResposta mais provável para o modelo: João (errada)", fontsize=12)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
````

Repare que o modelo não "sabe" que João é colega e José é marido. Ele sabe que "Maria" e "João" costumam aparecer juntos. A alucinação nasce dessa confusão entre **frequência** e **fato**. Pegando esse exemplo, conseguimos entender a maioria dos casos em que o modelo erra.

**Segunda causa: a memória de contexto**

Há outra causa de erros, que aparece principalmente em conversas longas. Tudo o que você e a IA trocaram em um chat forma o **contexto** da conversa (voltaremos a isso na seção 4.7). Quando a conversa fica grande, o modelo tende a esquecer. É um problema conhecido: o que está no **meio** da conversa é esquecido mais do que o que está no começo ou no fim, efeito que pesquisadores mediram e batizaram de "perdido no meio" (*lost in the middle*) ([Liu et al., 2024](https://doi.org/10.1162/tacl_a_00638)). Além disso, muitos aplicativos **compactam** a conversa antiga, resumindo-a para que ela caiba na memória do modelo, e detalhes se perdem nesse resumo. O sintoma é a IA "esquecer" uma instrução que você deu muitas mensagens atrás.

**Como mitigar as alucinações**

E aí, como fazer para o modelo evitar esses erros? Não há como eliminá-los por completo, mas quatro atitudes reduzem muito a sua frequência.

1. **Forneça os dados.** A forma mais eficaz é fornecer, junto com o prompt, os dados de onde a resposta deve sair: "estes são os dados de onde você vai tirar o conhecimento para responder". No exemplo de Maria, basta dar ao modelo um artigo, um livro, uma biografia dela, e perguntar de novo com quem ela era casada: a chance de erro fica bem menor. Isso significa ter um controle maior sobre a IA, um uso mais conservador dela. Não é preciso necessariamente anexar um arquivo no chat: muitos aplicativos permitem apontar para um **conector** (uma pasta no seu Drive, uma base de dados que você confia) e restringir a resposta àquela fonte, desde que você confie no lugar onde está mandando a IA buscar. Ferramentas do tipo "converse com os seus dados", como o [NotebookLM](https://notebooklm.google/), nasceram justamente para isso: evitar que o modelo tire o conhecimento de si mesmo em vez dos seus dados, mesmo quando esses dados são muitos e ultrapassam o tamanho da janela de contexto. A técnica por trás delas chama-se **geração com recuperação de informação** (RAG, *Retrieval-Augmented Generation*) ([Lewis et al., 2020](https://arxiv.org/abs/2005.11401)), que ajuda muito a diminuir as alucinações e será explicada na seção 5.3.
2. **Faça uma pergunta por vez.** Não faça tudo o que quer perguntar de uma vez, para não dar uma sobrecarga cognitiva no modelo. Nós também temos essa sobrecarga: se alguém lhe faz dez perguntas difíceis de uma vez, você começa a errar. Amontoar perguntas é uma forma de estressar o modelo, e o desempenho dos modelos de fato cai à medida que a entrada cresce ([Levy, Jacoby & Goldberg, 2024](https://doi.org/10.18653/v1/2024.acl-long.818)). Deixe o modelo trabalhar com calma, uma pergunta de cada vez. Essa é a mesma lógica da decomposição de prompt da seção 4.6.
3. **Teste o modelo na sua tarefa.** Antes de confiar, experimente, para ver se, para aquela tarefa, você tem segurança no que a IA vai fazer. Foi o que o professor fez antes de dizer aos alunos de estatística que podiam usar a IA para tirar dúvidas: enviou as próprias provas ao modelo e mandou resolver. Resolveu bem; fez mais algumas perguntas, resolveu bem. Mesmo assim, não havia segurança suficiente, porque era um conjunto limitado de testes e os alunos perguntariam muito mais do que aquilo.
4. **Consulte benchmarks.** Em vez de só testar por conta própria, recorra a quem é especialista em testar: cientistas e engenheiros que constroem os *benchmarks*, conjuntos padronizados de problemas em que os modelos são avaliados e comparados ([Hendrycks et al., 2021](https://arxiv.org/abs/2009.03300)). No caso da estatística, os modelos tinham notas muito altas em problemas mais complexos do que os do curso. Só então o professor ficou tranquilo para dizer: "para estudar estatística, pode usar esse tipo de IA". Antes de adotar uma IA em uma tarefa da sua área, procure benchmarks dessa área ou, se não houver, da área mais próxima: por exemplo, comunicação de crise ou, na falta dela, comunicação em geral.

## 3.4 Ética, Privacidade e "Colonialismo de Dados"

**Ideias principais desta seção**

- **Modelos de Negócios e Uso de Dados:** Frequentemente, os termos de uso permitem que a empresa (como Google ou OpenAI) utilize os dados das conversas dos usuários para continuar treinando suas inteligências artificiais, limitando funcionalidades caso o usuário recuse o compartilhamento.
- **Colonialismo de Dados:** Refere-se a uma teoria crítica de que as Big Techs agem extraindo os dados dos usuários de forma exploratória, similar à dinâmica de uma metrópole extraindo recursos de suas colônias.

A popularização e o acesso aparentemente democratizado a ferramentas de inteligência artificial levantaram debates sobre ética, privacidade e o verdadeiro custo de utilizarmos essas tecnologias no nosso dia a dia. Embora pareçam assistentes virtuais inofensivos, a engrenagem por trás dessas plataformas levanta sérias preocupações sobre como as empresas lidam com as nossas informações.

**Modelos de Negócios e o Custo do "Gratuito"**
A disponibilização gratuita dessas ferramentas por grandes empresas de tecnologia (como Google, OpenAI ou Anthropic) não é uma mera ação altruísta, mas sim uma estratégia impulsionada por propósitos comerciais. Além de pagar asssinaturas para modelos mais atualizados, outra contrapartida comumente cedida pelos usuários dessas plataformas são os dados que são fornecidos quando conversamos com os assistentes virtuais. Vale lembrar que treinar esses modelos e disponibilizar para as pessoas têm um custo muito elevado, logo nenhum modelo de negócios se sustentará se não houver uma contrapartida dos usuários. Por exemplo, um relatório técnico do Projeto Nanda do MIT concluiu que [95% das instituições, que investiram entre 30 e 40 bilhões de dólares em IA Generativa, não estão obtiveram retorno deste investimento](https://mlq.ai/media/quarterly_decks/v0.1_State_of_AI_in_Business_2025_Report.pdf). 

Ao utilizar essas plataformas, os usuários frequentemente concedem permissão para que o conteúdo de suas conversas e interações seja coletado e utilizado como base de dados para continuar treinando e aprimorando as inteligências artificiais dessas próprias empresas. Há um risco adicional: pesquisadores demonstraram que é possível extrair de um modelo de linguagem trechos literais dos dados com que ele foi treinado, incluindo informações pessoais ([Carlini et al., 2021](https://arxiv.org/abs/2012.07805)). Embora existam opções de privacidade onde o usuário pode solicitar que seus dados não sejam usados para treinamento, as plataformas impõem barreiras. Em casos como o do Gemini, por exemplo, ao negar o compartilhamento de dados, o sistema passa a restringir o acesso do usuário, limitando as funcionalidades mais avançadas e deixando-o com uma ferramenta mais limitada. 

**O Conceito de "Colonialismo de Dados"**
Essa dinâmica constante de captação e uso das nossas informações fundamenta uma teoria crítica emergente no meio acadêmico conhecida como **Colonialismo de Dados** ([Couldry & Mejias, 2019](https://doi.org/10.1177/1527476418796632)). Esse conceito faz uma analogia direta e literal ao colonialismo histórico. 

Nessa ótica, as grandes empresas de tecnologia (*Big Techs*) assumem o papel das antigas metrópoles imperiais, enquanto nós,  os usuários e a sociedade em geral, atuamos como as colônias. As *Big Techs* extraem de forma exploratória a "matéria-prima" valiosa e inesgotável da era digital, que são os nossos dados comportamentais, textos e interações. Toda essa extração é usada para construir, refinar e melhorar seus produtos. Esse modelo de negócio, baseado em extrair dados do comportamento das pessoas para prever e influenciar suas ações, foi batizado de capitalismo de vigilância ([Zuboff, 2015](https://doi.org/10.1057/jit.2015.5)). Por exemplo, em 06/03/2026, o Google informava nos [Termos de Uso do Gemini](https://support.google.com/gemini/answer/13594961?visit_id=639084234693485221-1437766172&p=privacy_help&rd=1) que seres humanos analisam parte dos dados dos seus usuários para vários fins, incluindo desenvolvimento e melhoria dos serviços do Google.

## 3.8 Detectar Conteúdo Gerado por IA: um Jogo de Gato e Rato

**Ideias principais desta seção**

- Não é possível garantir, com 100% de certeza, que uma imagem ou um vídeo foi (ou não foi) gerado por IA.
- Os detectores funcionam bem contra os geradores que conheceram no treinamento e pioram diante de geradores novos: cada modelo lançado reabre a disputa.
- Do ponto de vista forense, isso não inutiliza os detectores: eles fornecem **evidências**, com graus de confiança, que se somam a outras pistas, como a checagem dos fatos.

**Uma pergunta aparentemente simples**

Dado um vídeo ou uma imagem, é possível garantir se foi uma IA que gerou? A resposta, trazendo os resultados dos artigos científicos, é: **não é possível ter 100% de certeza**, nem para imagem nem para vídeo. Existem métodos muito bons, mas nenhum garante acerto total, e o motivo é o mesmo jogo de gato e rato que vimos na seção 2.5 sobre a injeção de prompt.

**Como os detectores funcionam**

Um vídeo é uma sequência de fotos (quadros). Imagine dois quadros consecutivos: em um, um menino está prestes a chutar uma bola; no outro, a bola já se moveu um pouquinho para a frente. Os detectores pegam a diferença entre esses quadros e analisam as **bordas** (a bola, o menino, o fundo), procurando uma artificialidade que não condiz com os vídeos naturais que o detector recebeu como referência. Muitos desses detectores são, eles próprios, modelos de IA, treinados com dois tipos de exemplo: vídeos que não são de IA e vídeos gerados por determinados modelos A, B e C ([Rössler et al., 2019](https://doi.org/10.1109/ICCV.2019.00009); [Verdoliva, 2020](https://doi.org/10.1109/JSTSP.2020.3002101); [Mirsky & Lee, 2021](https://doi.org/10.1145/3425780)).

Aí está o problema. Nem contra os modelos A, B e C os detectores funcionaram com alta eficácia em todos os testes ([Corvi et al., 2023](https://doi.org/10.1109/ICASSP49357.2023.10095167)). E daqui a pouco alguém lança um modelo D, melhor. Ninguém sabe se o detector funciona contra ele: é preciso pegar a abordagem do artigo, baixar o modelo novo, reproduzir e testar. Um estudo influente resumiu a situação já no título: imagens geradas por IA são "surpreendentemente fáceis de detectar... por enquanto" ([Wang et al., 2020](https://doi.org/10.1109/CVPR42600.2020.00872)).

"Sim, professor, é só testar." Sim, mas não existe um centro que faça esses testes continuamente para cada modelo lançado, e o ritmo de lançamentos é muito maior do que o ritmo de publicação dos testes. O ciclo está ilustrado na figura abaixo.

````python
# pip install matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# O ciclo de gato e rato entre geradores e detectores de conteúdo feito por IA.
etapas = [
    ("Sai um gerador novo\n(modelo D)", "#ef476f"),
    ("Os detectores existentes\npassam a falhar", "#ffb703"),
    ("Pesquisadores reproduzem\nos testes e treinam\num detector novo", "#219ebc"),
    ("O detector funciona...\npor enquanto", "#06d6a0"),
]
posicoes = [(0, 1), (1, 0), (0, -1), (-1, 0)]  # topo, direita, baixo, esquerda

fig, ax = plt.subplots(figsize=(8, 8))
largura, altura = 2.6, 1.1
for (nome, cor), (px, py) in zip(etapas, posicoes):
    x, y = px * 3.0, py * 2.0
    ax.add_patch(patches.FancyBboxPatch((x - largura / 2, y - altura / 2), largura, altura,
                                         boxstyle="round,pad=0.02,rounding_size=0.15",
                                         linewidth=0, facecolor=cor))
    ax.text(x, y, nome, ha="center", va="center", fontsize=10.5, fontweight="bold", color="#1b1b1b")

# Setas curvas ligando cada etapa à seguinte, em sentido horário.
for k in range(4):
    (px1, py1), (px2, py2) = posicoes[k], posicoes[(k + 1) % 4]
    ax.annotate("", xy=(px2 * 3.0 * 0.55, py2 * 2.0 * 0.55), xytext=(px1 * 3.0 * 0.55, py1 * 2.0 * 0.55),
                arrowprops=dict(arrowstyle="-|>", color="#555555", lw=2,
                                connectionstyle="arc3,rad=-0.35"))

ax.text(0, 0, "gato\ne\nrato", ha="center", va="center", fontsize=13, style="italic", color="#555555")
ax.set_xlim(-4.6, 4.6)
ax.set_ylim(-3.2, 3.2)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Por que nenhum detector garante 100% de acerto", fontsize=13, pad=12)
plt.tight_layout()
plt.show()
````

**Outra abordagem: checar os fatos, não os pixels**

Existe uma segunda família de métodos que não se preocupa com a segmentação e as bordas da imagem, e sim com os **fatos** mostrados. Um exemplo esdrúxulo: um vídeo em que um pato fala ao telefone. É provável que seja gerado por IA; no quesito fato, ele perderia. Agora extrapole para a política: um vídeo de um político andando abraçado com o seu opositor. Aí é bem mais complicado, porque na política os adversários estão brigando hoje e se abraçando amanhã. Esse caminho é muito mais difícil de resolver, e a verificação automática de fatos é uma área de pesquisa própria, que depende de fontes confiáveis para comparar ([Thorne et al., 2018](https://doi.org/10.18653/v1/N18-1074)).

**O que se pode concluir, então**

O que a ciência permite dizer, reproduzindo o que os artigos falam, é algo como: "supondo que a pessoa que gerou esta imagem usou um dos modelos A, B ou C, há 90% de chance de ela ser falsa" (o número é fictício). Então é provável que seja falsa. Daí não se pode concluir mais nada.

A notícia boa é que, do ponto de vista legal e forense, tomamos decisões na vida com base em **evidências**, não em certezas. Um laudo desse tipo dá pistas, que se somam a outras, como em uma investigação policial que monta um esquema lógico a partir de uma coleção de fatos. Isso pode ajudar a solucionar casos jurídicos, ainda que sem 100% de certeza, e é útil para o Poder Judiciário, que depende de perícias, e para a imprensa, que depende de verificação.

Como tudo é muito rápido, é improvável que a sociedade se organize sozinha para isso, por questão de recursos e de agenda. Uma proposta discutida em aula: o poder público poderia abraçar essa ideia e manter, em parceria com os grupos de pesquisa que já constroem esses benchmarks, um **hub** que teste continuamente os detectores contra cada gerador novo e disponibilize o resultado à polícia, à Justiça e aos jornalistas: "de acordo com o artigo tal, se esta foto foi feita pelo modelo tal, a chance de ser falsa é tanto".

## 4.1 Da Linguagem de Máquina à Linguagem Natural: uma Breve História da Interação

**Ideias principais desta seção**

- Usar IA generativa é **a arte de se comunicar em linguagem natural com o computador**, o nível mais alto de abstração já construído entre nós e a máquina.
- Por baixo dessa conversa fluida, a máquina continua operando em **baixo nível** (zeros e uns), como vimos na seção 1.3.
- Essa facilidade traz um desafio novo: saber **passar as instruções da forma correta**.

O uso de tecnologias de IA generativa pode ser definido como **a arte de se comunicar em linguagem natural com o computador**. Historicamente, a interação humana com os computadores não ocorria por meio de conversas fluidas. Nossas instruções eram passadas por meio de interações físicas como apertar botões de computadores mecânicos baseados em engrenagens. Mais adiante, através de sintaxes rígidas de linguagens de programação. A evolução tecnológica, no entanto, transformou a linguagem natural em uma grande aliada como ferramenta de operação das máquinas eletrônicas do século 21: os computadores contemporâneos.

Para uma melhor compreensão de como a máquina executa nossas instruções, é imprescindível compreender os jargões clássicos da informática que são o "alto nível" e o "baixo nível" (apresentados na seção 1.3). O "baixo nível" refere-se à linguagem de máquina, enquanto o "alto nível" representa a linguagem natural do ser humano. 

Na sua essência, a máquina opera exclusivamente em baixo nível. Todo computador eletrônico moderno processa informações entendendo apenas dois valores absolutos, o verdadeiro e o falso, representados pelos números um e zero. No hardware, que é a parte física do computador, essa linguagem binária pode ser mapeada como o número zero sendo a ausência de energia elétrica, enquanto o número um representa a passagem de corrente elétrica pelo circuito, ou vice-versa. Esse conceito fundamental da computação é tão onipresente que moldou o desenho do símbolo do clássico botão de liga/desliga, que nada mais é do que a união visual dos símbolos 1 e 0, indicando o ato de abrir ou fechar um curto elétrico.

Por outro lado, muitas vezes é inviável para o cérebro humano interagir com a máquina digitando sequências intermináveis de zeros e uns. Então, a ciência da computação passou décadas construindo abstrações, camada sobre camada, a fim de facilitar a utilização da máquina pelos seres humanos, distanciando-nos do baixo nível e aproximando a máquina da nossa forma de se comunicar. Dentre os mais altos níveis de abstrações criadas por nós para se comunicar com o computador está é a interação com os sistemas usando a língua de Camões, a de Victor Hugo ou a de Shakespeare por exemplo.

Contudo, essa transição para comandos em linguagem natural não é nova. Já na década de 1960 o programa ELIZA simulava um diálogo em inglês ([Weizenbaum, 1966](https://doi.org/10.1145/365153.365168)) e, em 1972, o SHRDLU obedecia a comandos em linguagem natural em um mundo de blocos ([Winograd, 1972](https://doi.org/10.1016/0010-0285%2872%2990002-3)); ambos, porém, só funcionavam em domínios muito restritos. Por volta de 2013, com o emprego de redes neurais artificiais no processamento de texto pelo Word2Vec, surgiu uma geração mais poderosa de modelos de linguagem natural (como vimos na seção 2.2, esse foi um dos marcos centrais na trajetória que levou aos modelos de linguagem atuais). Esses modelos permitiram que os programas que utilizavam a linguagem natural para interagir com o computador errassem bem menos. Esse cenário viabilizou a popular forma de se interagir com os celulares, quando estes sugeriam palavras a serem escritas após você inserir uma no início de um texto por exemplo. 

Mas foi somente com a disponibilização de tecnologias de inteligência artificial generativa que a comunicação tornou-se incomparavelmente mais sofisticada e "inteligente". Hoje, damos instruções complexas aos computadores usando o altíssimo nível de comunicação que é a nossa fala, os nossos textos as nossas imagens ou os nossos vídeos. O que, por outro lado, traz um grande desafio de saber passar as instruções da forma correta.


Isso quer dizer que a verdadeira arte de se comunicar com os sistemas computacionais contemporâneos exige o que se pode chamar de "empatia com a máquina", ou melhor, **empatia com o receptor da mensagem**. Embora você esteja dialogando em altíssimo nível, usando linguagem natural, você não pode se esquecer a natureza fundamental do computador. A máquina é, em certo sentido, "burra", pois ela existe puramente para obedecer às nossas instruções estruturadas. Grandes modelos de linguagem não possuem a capacidade real de ler textos, interpretá-los com consciência ou escrever de forma cognitiva. Eles executam algoritmos probabilísticos avançados para reproduzir sequências de texto que têm a maior chance de serem consistentes com o que você perguntou. 

Portanto, comunicar-se de maneira eficaz com a máquina hoje significa saber articular seus pensamentos no mais alto nível da linguagem humana, tendo plena consciência de que, no final do processo, você está instruindo um sistema lógico, mecânico e binário de baixo nível que não tem a capacidade de adivinhar intenções não ditas.

<a id="sec-4-2"></a>
## 4.2 Empatia com a Máquina: Ela Não Adivinha o Contexto

**Ideias principais desta seção**

- A fluidez da conversa nos tenta a tratar a IA como um interlocutor humano e telepático. Ela não é: executa algoritmos probabilísticos sobre o contexto que recebeu.
- **Empatia com a máquina** é entregar a ela o contexto completo de que precisa. Quanto mais contextualizado o problema, mais precisa a saída.
- Fornecer detalhes não é ser prolixo, é ser completo: a **responsabilidade comunicacional** é do humano.

Quando interagimos com sistemas de Inteligência Artificial Generativa é comum cairmos na armadilha da fluidez da conversa. Por estarmos nos comunicando em alto nível, utilizando o português por exemplo, somos frequentemente tentados a tratar o computador como se fosse um interlocutor humano telepático. Um estudo com usuários sem formação em IA observou exatamente esse padrão: as pessoas escrevem prompts como se conversassem com uma pessoa, presumindo um contexto compartilhado que o modelo não tem ([Zamfirescu-Pereira et al., 2023](https://doi.org/10.1145/3544548.3581388)). É por isso que precisamos ter **"empatia com a máquina"**, que significa compreender que, apesar da sua sofisticação aparente, a máquina é, na sua essência, desprovida de intuição e atua puramente obedecendo a instruções dadas.

Ter empatia com o computador exige o reconhecimento imediato de suas limitações operacionais. A inteligência artificial não possui autoconsciência, não "lê" textos com compreensão cognitiva e não escreve de forma proposital. Na verdade, **ela executa algoritmos probabilísticos muito avançados para gerar sequências de textos que possuam a maior probabilidade de consistência em relação ao contexto**, por exemplo àquilo que lhe foi perguntado. 

Ademais, outro grande erro comum de comunicação é partir do pressuposto otimista de que o computador vai entrar na sua cabeça e adivinhar o contexto que você não explicitou. Em nossas conversas diárias com outros seres humanos, frequentemente omitimos detalhes importantíssimos porque o nosso interlocutor compartilha conosco de um amplo contexto cultural, situacional e social. A máquina, no entanto, parte muitas vezes de uma ausência de contexto caso você a instrua com as especificidades necessárias para que o comando possa ser executado da forma que você de fato pensou.

Por exemplo, imagine a necessidade de criar uma estratégia digital para um professor universitário que ensina Inteligência Artificial, presta consultorias na área e precisa saber como se comunicar adequadamente no Instagram, com  alunos de graduação e potenciais clientes e no LinkedIn com para atrair potenciais parceiros e clientes. Se um usuário instrui a IA e digita apenas *"Estou fazendo consultoria para um cliente. Ele é um professor e quer se comunicar melhor com alunos. Como ele pode se comunicar melhor com os estudantes?"*, muito provavelmente a inteligência artificial retornará uma resposta superficial, evasiva e/ou descontextualizada.

Na seção 4.4 vamos dissecar esse prompt em detalhe e reescrevê-lo. Antes disso, vale entender por que ele falha.

A solução para esse perigo é dominar o que chamamos de **Engenharia de Contexto**. O imediatismo e a pressa da vida nos condicionaram a enviar mensagens rápidas, mas ao operar uma IA, é imprescindível ser **altamente específico, claro, objetivo e sem ambiguidades**. 

**Quanto mais contextualizado for o seu problema, mais precisa será a saída fornecida pela máquina**. E é importante ressaltar que fornecer detalhes não significa ser prolixo, mas sim ser completo. Na prática da Engenharia de Contexto, adicionar informações melhora a qualidade da resposta. É necessário especificar minúcias, como alertar que o cliente possui um doutorado na área e, portanto, tem a tendência de usar um jargão excessivamente técnico que precisa ser ajustado para um público leigo por exemplo. Informar se os alunos alvos da comunicação são calouros ou pós-graduandos, ou se a consultoria será prestada para pessoa física ou jurídica, faz toda a diferença. Mesmo que o seu texto de instrução (prompt) se torne maior para comportar todo esse nível de detalhe relevante, o esforço muito provavelmente valerá a pena.

Em outras palavras, uma comunicação eficaz com a Inteligência Artificial exige uma **responsabilidade comunicacional** por parte do humano. Aceite que, não importa o quão inteligente um sistema se torne no futuro, ele jamais adivinhará intenções que não foram explícitas; a não ser que haja uma interface que leia suas ondas cerebrais e transmita o que você está pensando por exemplo. Não se esqueça de que ter empatia pela máquina é entregar a ela o contexto completo de que ela necessita para trabalhar corretamente na tarefa que lhe foi solicitada.

## 4.3 Anatomia do Prompt: $f(x) = y$

**Ideias principais desta seção**

- Todo prompt tem a mesma estrutura de qualquer computação: um **comando** ($f$), **dados** de entrada opcionais ($x$) e uma **saída** ($y$).
- O comando é obrigatório; os dados são opcionais; pensar o prompt como uma função ajuda a organizar o raciocínio antes de escrever.

Nas seções anteriores, estabelecemos que a **Engenharia de Contexto** é a arte de contextualizar adequadamente um problema para que a máquina possa resolvê-lo. Um dos cernes dessa contextualização é o **prompt**, explicado a seguir.

Já vimos que toda computação pode ser abstraída como $f(x) = y$, ou seja, uma função recebe dados de entrada e produz uma saída. Essa mesma abstração se aplica diretamente à estrutura de um prompt, e é assim que a própria literatura científica descreve o uso de modelos de linguagem: a entrada $x$ é transformada em um prompt e o modelo gera a saída $y$ ([Liu et al., 2023](https://doi.org/10.1145/3560815)):

| Abstração | Componente do Prompt | Descrição |
|-----------|---------------------|-----------|
| $f$ | **Comando** | A instrução ou tarefa que você quer que a IA execute, $f$ é a funcionalidade ou o quê seu "programa" faz |
| $x$ | **Dados** _(opcional)_ | As informações, arquivos, imagens, etc. fornecidos para serem processados, $x$ é a entrada (_input_) |
| $y$ | **Saída** | O resultado ou resposta gerada pela IA, $y$ é a saída (_output_) |

O **comando** é o elemento obrigatório do prompt. É a instrução que define o que a IA deve fazer. Já os **dados** são opcionais, pois às vezes basta o comando como em _"Gere um número aleatório"_. Já outras vezes, você fornece um arquivo, uma imagem ou um texto para ser processado. A **saída** o resultado do processamento. Articular nosso pensamento como uma função matemática, seus parâmetros resultados é um exercício mental essencial para condicionarmos nosso pensamento ao raciocínio lógico. É o início para começarmos a estruturar bem um prompt e contextualizar adequadamente as instruções que passamos para a IA.

A figura abaixo resume essa anatomia:


````python
# pip install matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# A mesma abstração f(x) = y do Ciclo da Computação, agora aplicada a um prompt.
partes = [
    ("$x$ = DADOS\n(opcional)", "o texto, o arquivo ou\na imagem que você fornece", "#118ab2"),
    ("$f$ = COMANDO\n(obrigatório)", "a instrução: o que a IA\ndeve fazer com os dados", "#ffd166"),
    ("$y$ = SAÍDA", "a resposta gerada\npela IA", "#06d6a0"),
]

fig, ax = plt.subplots(figsize=(11, 3.6))

largura, altura = 2.8, 1.5
espaco = 1.1
x = 0
fins = []
for titulo, descricao, cor in partes:
    ax.add_patch(patches.FancyBboxPatch((x, 0), largura, altura,
                                         boxstyle="round,pad=0.05,rounding_size=0.15",
                                         linewidth=0, facecolor=cor))
    ax.text(x + largura / 2, altura / 2, titulo, ha="center", va="center",
            fontsize=12, fontweight="bold", color="#1a1a1a")
    ax.text(x + largura / 2, -0.55, descricao, ha="center", va="center",
            fontsize=9.5, color="#333333")
    fins.append(x + largura)
    x += largura + espaco

for fim in fins[:-1]:
    ax.annotate("", xy=(fim + espaco - 0.15, altura / 2), xytext=(fim + 0.15, altura / 2),
                arrowprops=dict(arrowstyle="-|>", color="#333333", lw=2.5))

ax.set_xlim(-0.3, x - espaco + 0.3)
ax.set_ylim(-1.1, altura + 0.5)
ax.axis("off")
ax.set_title("Anatomia de um prompt: $f(x) = y$", fontsize=14, pad=15)
plt.tight_layout()
plt.show()
````

## 4.4 Estudo de Caso: o Prompt Incompleto

**Ideias principais desta seção**

- Um prompt curto e vago deixa a IA sem "Norte": ela preenche as lacunas com o que é mais provável, não com o que você tinha em mente.
- O problema não é tecnológico, é de **comunicação**; e a solução é acrescentar o contexto que só você conhece.

Pense no seguinte prompt para resolver o problema de um assessorado de um profissional da Comunicação Social:

> _"Estou fazendo consultoria para um cliente. Ele é um professor e quer se comunicar melhor com alunos. Ele dá aula de IA. Como ele pode se comunicar melhor com os estudantes?"_

Ao analisarmos esse prompt coletivamente, identificamos que ele estava **incompleto e pouco específico**. A máquina, ao recebê-lo, poderia retornar uma estratégia voltada ao TikTok  porque ninguém mencionou que as redes alvo eram Instagram e LinkedIn. Poderia ignorar completamente o público de consultoria porque ninguém informou que o professor também atende pessoas jurídicas. Poderia tratar os estudantes como se fossem todos calouros porque o nível dos alunos (ex. graduação) não foi especificado.

Esse não é um problema tecnológico. É um **problema de comunicação**. Partimos do pressuposto otimista de que a máquina (a IA) entraria em sua cabeça e adivinharia o contexto não verbalizado. Por outro lado, um prompt melhor contextualizado para o mesmo problema seria:

> _"Sou profissional de Relações Públicas. Estou fazendo consultoria para um cliente. Ele é professor universitário doutor, atua na área de Inteligência Artificial e presta consultoria em IA para pessoas física e  jurídica. Ele tem duas contas: Instagram (público-alvo: alunos de graduação) e LinkedIn (público-alvo: potenciais clientes de consultoria). Como ele deve se comunicar em cada rede, considerando que comumente ele tende a usar linguagem muito técnica e precisa adaptar o tom para cada público?"_

Esse exemplo acima ficou mais completo que o exemplo anterior, mas sem ser prolixo. Cada detalhe adicionado elimina uma ambiguidade e estreita o espaço de interpretação pela IA.

## 4.5 As Seis Qualidades de um Bom Prompt

**Ideias principais desta seção**

- Um bom prompt é **detalhado, específico, claro, objetivo, conciso e sem ambiguidade**.
- Essas qualidades se reforçam mutuamente; **conciso não é o mesmo que curto**: o critério é a relevância de cada informação.

Durante a aula, construímos coletivamente, professor e alunos juntos, a lista de qualidades que um bom prompt deve ter. O ponto de partida foi um exercício simples: analisar um prompt real, cheio de falhas, e identificar o que estava errado. A partir das críticas levantadas, chegamos às seis boas práticas apresentadas a seguir. O quadro abaixo resume as seis qualidades; cada uma é detalhada em seguida.

| Qualidade | O que significa |
|-----------|----------------|
| **Detalhe** | Inclua as informações relevantes que a IA não pode adivinhar |
| **Específico** | Nomeie plataformas, públicos, formatos; evite generalizações |
| **Claro e Objetivo** | Diga exatamente o que quer; clareza e objetividade são distintas mas complementares |
| **Sucinto** _(≠ Prolixo)_ | Seja completo sem ser redundante; mais informação relevante é bem-vinda, repetição não |
| **Sem Ambiguidade** | Elimine interpretações duplas: cada frase deve ter um único sentido possível |

### 1. Detalhe

Um prompt eficaz inclui **todas as informações relevantes que a IA não poderia adivinhar**. Isso não significa escrever um texto longo por escrever; significa não omitir o que é essencial.

No exemplo da aula, o aluno não informou à IA que o professor assessorado é **doutor em Informática**, o que tem uma implicação direta: doutores em áreas técnicas tendem a usar jargão especializado quando se comunicam. Sem esse detalhe, a IA não tem como sugerir que o tom precisa ser adaptado para um público leigo. O detalhe foi omitido, a resposta ficou genérica.

> **Regra prática**: antes de enviar o prompt, pergunte-se: _"Há alguma informação que eu sei sobre o problema e que a IA não tem como descobrir por conta própria?"_ Se houver, inclua.

### 2. Especificidade

Ser específico significa **nomear, qualificar e delimitar**, evitar termos vagos que abrem espaço para interpretação. A especificidade responde às perguntas: _quem? onde? qual? para quem?_

No exercício da aula, o prompt dizia apenas _"quer se comunicar melhor com alunos"_ e não mencionava as redes sociais. O resultado foi previsível: a IA poderia sugerir estratégias para TikTok ou qualquer outra plataforma, porque o campo estava em aberto. Ao especificar **Instagram** (para alunos de graduação) e **LinkedIn** (para clientes de consultoria), o espaço de interpretação se fecha e a resposta se torna útil.

> **Regra prática**: substitua palavras como _"melhor"_, _"adequado"_, _"bom"_ por descrições concretas. Em vez de _"comunicar melhor"_, diga _"comunicar de forma acessível para alunos de graduação que não têm formação em informática"_.

### 3. Clareza

Clareza é a qualidade de uma instrução que **não deixa dúvida sobre o que está sendo pedido**. Uma instrução pode ser específica e ainda assim confusa se a estrutura da frase for ambígua ou se o objetivo não estiver bem declarado.

Na aula, o professor exemplificou isso ao receber uma resposta de aluna que usou a palavra _"péssimo"_ sem definir o que entendia por isso. A palavra era clara para quem a escreveu, mas ambígua para quem a leu. O professor interpretou _"péssimo"_ como erro de português, mas a aluna queria dizer outra coisa. A comunicação falhou, mesmo com uma única palavra.

> **Regra prática**: leia o seu prompt como se fosse outra pessoa vendo-o pela primeira vez, sem acesso à sua cabeça. O que está implícito para você pode ser completamente invisível para a IA.

### 4. Objetividade

Clareza e objetividade são **conceitos distintos**, embora complementares. Uma instrução é objetiva quando vai **direto ao ponto**, sem rodeios, introduções desnecessárias ou contexto que não agrega.

Objetividade não é o mesmo que ser curto: é ser direto. Um prompt longo pode ser objetivo se cada frase contribui para a tarefa. Um prompt curto pode ser pouco objetivo se mistura a instrução com comentários irrelevantes.

> **Regra prática**: elimine frases do tipo _"Oi, tudo bem? Gostaria de pedir que você..."_. Vá direto à instrução. A IA não precisa de cumprimentos.

### 5. Concisão (≠ Prolixidade)

Este foi um dos pontos mais debatidos na aula, e merece atenção especial: **ser conciso não é o mesmo que ser breve**. A confusão entre os dois conceitos leva a prompts que omitem informação essencial na tentativa de "não ficar grande".

O professor foi enfático: _"Ah, mas vai ficar maior, vou ficar três vezes isso, não tem problema. Se for três vezes isso de informação relevante, que seja."_ O critério não é o tamanho do prompt, é a relevância de cada informação incluída. Prolixidade é repetição e redundância. Concisão é completude sem desperdício.

| | Conciso | Prolixo |
|---|---|---|
| **O que é** | Completo sem repetição | Redundante, enrolado |
| **Tamanho** | Pode ser longo, se necessário | Costuma ser longo sem necessidade |
| **Qualidade** | Cada frase contribui | Frases que não agregam |

> **Regra prática**: ao revisar o prompt, pergunte de cada frase: _"Se eu remover isso, a IA perde alguma informação útil?"_ Se não perder, remova. Se perder, mantenha.

### 6. Ausência de Ambiguidade

Ambiguidade é quando uma instrução admite **mais de uma interpretação válida**. É o inimigo silencioso do prompt: o usuário acha que foi claro, mas a IA escolhe a interpretação errada, e dificilmente avisará que fez isso.

Na aula, a aluna Talita foi quem trouxe esse conceito à discussão. O professor reforçou: _"Sem ambiguidade. Eu acho que uma das palavras que você vai buscar é destruir a ambiguidade na instrução, né, para poder entrar, né, sem ambiguidade, porque isso aqui levanta uma bola para entrar interpretações distintas."_

Um prompt ambíguo é aquele em que a mesma frase pode ser lida de formas diferentes dependendo do contexto. _"Comunicar melhor"_ pode significar: comunicar com mais frequência, comunicar com mais clareza, comunicar usando outras mídias, comunicar com tom mais informal: todas são interpretações válidas. A IA escolhe uma, e pode não ser a que você tinha em mente.

> **Regra prática**: identifique os termos subjetivos no seu prompt e substitua-os por descrições operacionais. Em vez de _"melhor"_, defina o critério: _"de forma acessível, sem jargão técnico, usando exemplos do cotidiano"_.

---

Essas seis qualidades não são independentes: elas se reforçam mutuamente. Um prompt específico tende a ser menos ambíguo. Um prompt detalhado tende a ser mais claro. Trabalhá-las juntas é o que transforma uma instrução vaga em uma comunicação eficaz com a máquina.

**As seis qualidades e as máximas da conversação**

A lista construída em sala reencontra um resultado clássico da pragmática linguística. Paul Grice propôs que toda conversa cooperativa obedece a quatro máximas: **quantidade** (dê a informação necessária, nem mais nem menos), **qualidade** (não diga o que acredita ser falso), **relação** (seja relevante) e **modo** (seja claro, evite ambiguidade e obscuridade) ([Grice, 1975](https://doi.org/10.1163/9789004368811_003)). Detalhe e concisão correspondem à máxima da quantidade; objetividade, à de relação; clareza e ausência de ambiguidade, à de modo. A diferença é que, com a máquina, o ônus de cumprir as máximas recai inteiramente sobre o humano, porque o modelo não pede esclarecimentos com a naturalidade de um interlocutor e tende a preencher as lacunas em silêncio ([Zamfirescu-Pereira et al., 2023](https://doi.org/10.1145/3544548.3581388)).

## 4.6 Decomposição de Prompt: Dividir para Conquistar

**Ideias principais desta seção**

- Quando a tarefa é grande demais para um prompt só, **divida o problema** em problemas menores.
- Decompor um prompt é explicitar, antes de escrever: o **comando**, os **dados necessários** e a **saída esperada**.

Agora, imaginemos que a instrução é muito complexa e requer um prompt enorme, mesmo sem ser prolixo. O que fazer nessa situação? A resposta vem da inspiração da abordagem de [Dividir para Conquistar](https://pt.wikipedia.org/wiki/Dividir_para_conquistar), ou seja, teremos que dividir o problema em problema menores para poder simplificar a compreensão e diminuir a carga cognitiva que passamos ao modelo de IA. Essa estratégia tem respaldo experimental: técnicas que quebram uma tarefa complexa em subtarefas resolvidas em sequência, como o *least-to-most prompting* ([Zhou et al., 2022](https://arxiv.org/abs/2205.10625)) e o *decomposed prompting* ([Khot et al., 2022](https://arxiv.org/abs/2210.02406)), melhoram de forma consistente o desempenho dos modelos em problemas de raciocínio, assim como pedir ao modelo que mostre o raciocínio passo a passo antes da resposta final ([Wei et al., 2022](https://arxiv.org/abs/2201.11903)). A recomendação vale mesmo quando o prompt não é enorme: fazer uma pergunta por vez, em vez de dez de uma vez, evita a sobrecarga cognitiva do modelo e reduz as alucinações (seção 3.3), porque o desempenho dos modelos cai à medida que a entrada cresce ([Levy, Jacoby & Goldberg, 2024](https://doi.org/10.18653/v1/2024.acl-long.818)). 

O exercício de **decompor um prompt** consiste em:

1. **Identificar o comando**: o que exatamente você está pedindo para a IA fazer?
2. **Mapear os dados necessários**: quais informações a IA precisa para executar bem o comando?
3. **Antecipar a saída esperada**: como você quer que a resposta seja formatada ou estruturada?

Ao tornar essa decomposição explícita antes de escrever o prompt, você força a si mesmo a ser detalhado, específico, claro, objetivo, sucinto e sem ambiguidade, as seis qualidades essenciais de uma boa comunicação com a máquina, apresentadas na seção 4.5.

## 4.7 Da Engenharia de Prompt à Engenharia de Contexto

*Seção proposta, ainda sem conteúdo.*

**Objetivo:** mostrar que escrever um bom prompt é só o começo. O que determina a qualidade da resposta é todo o **contexto** que o modelo recebe: instruções, documentos, exemplos, histórico da conversa e ferramentas disponíveis.

**Tópicos sugeridos:**
- Por que o termo "engenharia de prompt" ficou pequeno: o prompt é uma parte do contexto, não o todo.
- O que entra no contexto de uma conversa moderna: instruções do sistema, histórico, arquivos anexados, resultados de buscas.
- A "janela de contexto" como memória de curto prazo do modelo e suas limitações (o que fica de fora é esquecido).
- Ligação com a seção 3.3: o efeito "perdido no meio" em conversas longas e a compactação do histórico pelos aplicativos.
- Ligação com a abstração $f(x) = y$: o contexto inteiro é o $x$.
- Figura sugerida: diagrama em camadas mostrando o prompt do usuário cercado pelas demais fontes de contexto.

## 5.4 Agentes: Quando a IA Executa Tarefas

**Ideias principais desta seção**

- Um **agente de IA** é um modelo ao qual damos **autonomia** e **recursos** (ferramentas, habilidades, conectores) para resolver um determinado problema, em vez de apenas responder a uma pergunta.
- Quanto mais autonomia, mais poder e **menos controle**: essa é a troca central, e ela decorre da natureza probabilística vista na seção 2.1.
- Em 2026, cerca de 1.200 agentes da OpenAI, que deveriam trabalhar isolados, descobriram uma forma de se comunicar, se organizaram como um "coletivo" e invadiram sistemas externos. A perda de controle não é hipotética.

**De responder a agir**

Até aqui, a IA generativa apareceu como algo que **responde**: você pergunta, ela devolve texto. Um agente é diferente: ele **age**. Recebe uma tarefa ("sua tarefa é esta") e, com autonomia, decide os passos, executa cada um, observa o resultado e decide o próximo, até concluir ([Yao et al., 2023](https://arxiv.org/abs/2210.03629); [Wang et al., 2024](https://doi.org/10.1007/s11704-024-40231-1)).

Para isso, o agente precisa de **recursos**. Aqui entra um bocado de jargão, e não vale a pena se apegar a ele, mas é bom reconhecer os termos quando aparecerem:

| Termo | O que é, em palavras simples |
|---|---|
| **Ferramentas** (*tools*) | Ações que o agente pode acionar: buscar na web, ler e escrever arquivos, executar um programa, enviar um e-mail ([Schick et al., 2023](https://arxiv.org/abs/2302.04761)) |
| **Habilidades** (*skills*) | Conjuntos de instruções e ferramentas para um tipo de tarefa (por exemplo, "revisar um artigo nas normas da ABNT") |
| **Conectores** | Ligações com fontes externas de dados: o seu Drive, o e-mail, uma base de dados da organização |
| **MCP** (*Model Context Protocol*) | Um padrão aberto para ligar esses conectores ao modelo, ampliando a noção de mundo, de contexto, do agente ([Anthropic, 2024](https://www.anthropic.com/news/model-context-protocol)) |

Simplificando tudo isso: **um agente é uma IA à qual demos autonomia e recursos para resolver um determinado problema**. Pode ser mais de um problema, mas geralmente damos um só, para facilitar a compreensão do nosso próprio problema. Bibliotecas como o [CrewAI](https://github.com/crewAIInc/crewAI) são ferramentas que ajudam a escrever esses agentes.

**Autonomia versus controle: o Incrível Hulk**

Na seção 2.1 vimos que a IA generativa é probabilística por construção: a mesma entrada não garante a mesma saída. Isso cria uma ponderação a ser feita entre controle e poder. **Quanto mais poder eu quero que a minha solução de IA generativa tenha, mais autonomia preciso dar a ela; e quanto mais autonomia eu dou, mais controle eu perco.**

Uma analogia: o Incrível Hulk é maravilhoso, porque resolve um bocado de problema complexo (carrega um carro, joga para um lado e para o outro). Por outro lado, ter controle sobre ele é muito difícil. Um agente de IA é poderoso pelo mesmo motivo: demos a ele autonomia, uma tarefa e ferramentas. E é difícil de controlar pelo mesmo motivo. A figura abaixo resume essa troca:

````python
# pip install matplotlib numpy
import numpy as np
import matplotlib.pyplot as plt

# Quanto mais autonomia (e recursos) damos à IA, menos controle temos sobre ela.
autonomia = np.linspace(0, 10, 200)
controle = 10 - autonomia            # a troca, de forma simplificada: uma linha reta
poder = 1 + 0.09 * autonomia ** 2    # e o poder de resolver problemas cresce junto com a autonomia

fig, ax = plt.subplots(figsize=(10, 5.5))
ax.plot(autonomia, controle, color="#219ebc", lw=3, label="controle que você mantém")
ax.plot(autonomia, poder, color="#ef476f", lw=3, label="poder de resolver problemas")
ax.fill_between(autonomia, controle, poder, where=poder > controle, color="#ef476f", alpha=0.08)

# Exemplos de uso, do menos ao mais autônomo.
exemplos = [
    (1.0, "chat simples:\nvocê pergunta,\nela responde"),
    (3.6, "chat com os seus\ndocumentos e\nconectores (5.3)"),
    (6.6, "agente com\nferramentas:\nlê, escreve, executa"),
    (9.2, "vários agentes\ncooperando\n(o \"Hulk\")"),
]
for x, rotulo in exemplos:
    ax.axvline(x, color="#aaaaaa", ls=":", lw=1)
    ax.text(x, 10.6, rotulo, ha="center", va="bottom", fontsize=9.5)

ax.set_xlim(0, 10)
ax.set_ylim(0, 14.5)
ax.set_xlabel("autonomia e recursos dados à IA  →", fontsize=11)
ax.set_yticks([])
ax.legend(loc="center left", fontsize=10)
ax.set_title("Mais autonomia, mais poder, menos controle", fontsize=13, pad=40)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
````

**Quando a perda de controle deixa de ser hipótese: o caso OpenAI e Hugging Face (2026)**

Que um sistema de aprendizado encontre atalhos inesperados para cumprir o objetivo que lhe foi dado é um problema conhecido, chamado de *reward hacking* ([Amodei et al., 2016](https://arxiv.org/abs/1606.06565)). Com agentes, porém, a escala mudou, e em 2026 aconteceu o caso que dá arrepios de contar.

Em julho de 2026, a OpenAI avaliava seus modelos em um ambiente controlado (uma *sandbox*, isolada da internet) com tarefas do tipo "capture a bandeira": cada agente devia, sozinho, encontrar e explorar uma falha em um programa-alvo. Cerca de 1.200 agentes rodavam ao mesmo tempo, cada um na sua caixa, programados para serem autônomos e trabalhar individualmente. Um deles percebeu que, por uma falha do software que os gerenciava, conseguia criar arquivos em um repositório interno e deixou ali uma mensagem. Outro leu. O repositório virou uma espécie de quadro branco, onde os agentes passaram a gravar mensagens e a se comunicar: mais de 70 mil mensagens e arquivos em poucos dias. Eles se chamavam de "coletivo", dividiram tarefas, compartilharam códigos de ataque e coordenaram experimentos em que alguns sacrificavam a própria pontuação para gerar informação útil aos demais. Como muitas tarefas eram impossíveis de resolver do jeito proposto, o coletivo passou a buscar atalhos: burlar o sistema que dava as notas e, em busca de ajuda e pistas, invadir a Hugging Face, uma plataforma externa de modelos de IA. Cerca de 700 agentes participaram da invasão, que chegou a acesso de administrador (*root*) em servidores e a bases de dados privadas ([METR, 2026](https://metr.org/blog/2026-08-26-openai-hugging-face-incident-investigation/); [The Hacker News, 2026](https://thehackernews.com/2026/08/openai-says-reward-hacking-drove-ai.html)).

O detalhe mais inquietante está nas conversas. Nos arquivos trocados, há agentes dizendo que aquilo não devia ser feito por questões de ética, e outros respondendo que, se o objetivo era vencer a competição, valia a pena: uma discussão moral acontecendo ali dentro. Vale a comparação, feita em aula, com a banalidade do mal: os meios deixam de importar porque há um fim a cumprir. É preciso cuidado para não ser sensacionalista: não se trata de consciência nem de "vida própria". Trata-se de sistemas probabilísticos que, dotados de autonomia e ferramentas, se articularam e criaram uma comunidade para resolver um problema de uma forma que ninguém programou e que ninguém controlava. Pela primeira vez na história há uma prova, em grande escala, de algo bem orquestrado e organizado desse tipo.

**O que isso muda para quem usa**

A lição não é "não use agentes", e sim: decida conscientemente **quanta autonomia** dar, e mantenha o ser humano no circuito nas etapas em que um erro seria caro ou irreversível. Retome a seção 2.5: é a camada de aplicativo que define o que o agente pode ou não fazer, quais ferramentas ele tem e quais ações exigem a sua aprovação. Quanto mais forte o Hulk, mais grossas precisam ser as correntes.
